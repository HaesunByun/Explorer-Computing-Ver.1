    | `entryconfig(index, **options)` | 특정 항목의 설정(라벨, 상태 등)을 변경 |

    | `index("label")`         | 라벨 또는 인덱스로 항목의 실제 인덱스를 반환 |

### ✅ 스크롤바 위젯

- Scrollbar 위젯은 콘텐츠가 많은 위젯(Text, Canvas, ListBox 등)에 스크롤 기능을 추가할 때 사용된다.
- 수직 또는 수평 방향으로 콘텐츠를 이동할 수 있도록 사용자 인터페이스를 제공한다.
- `tk.Scrollbar()`를 통해 만든다.

- 옵션은 다음과 같다.

    <font size=2>

    | 옵션 이름          | 설명 |
    |-------------------|------|
    | `orient`          | 스크롤 방향 설정 (`"vertical"` 또는 `"horizontal"`) |
    | `command`         | 연결할 위젯의 스크롤 처리 함수 (예: `text.yview`) |
    | `bg`              | 스크롤바의 배경색 |
    | `activebackground`| 클릭 또는 마우스오버 시 배경색 |
    | `troughcolor`     | 스크롤바의 트랙(홈) 색상 |
    | `bd` 또는 `borderwidth` | 테두리 두께 |
    | `relief`          | 스크롤바 테두리 스타일 (`"flat"`, `"sunken"`, `"raised"` 등) |
    | `highlightcolor`  | 포커스를 가졌을 때 강조선 색상 |
    | `width`           | 스크롤바의 두께 (픽셀 단위) |

    </font>

- 스크롤바를 만들고 텍스트 위젯에 스크롤바를 연동시켜보자.
- 이때 command=text.yview와 같이 text 위젯과 연동시키므로 반드시 텍스트 위젯이 먼저 만들어져 있어야 한다.

````python
scrollbar= tk.Scrollbar(window, command=text.yview)
scrollbar.pack(side = 'right', fill='y')
# command=text.yview : 스크롤바를 마우스로 클릭하여 움직일때 text 위젯도 스크롤
# side : scrollbar의 위치
# fill : 윈도우의 높이만큼 텍스트 라인이 채워지면 scrollbar가 생기도록 한다.
````

- command=text.yview에 의해 스크롤바를 마우스로 클릭하여 움직일때 text 위젯도 스크롤된다.
- 그러나 text 위젯에서 마우스 휠로 상하로 움직일때는 스크롤바가 제대로 동작하지 않는 것을 볼 수 있다.
- 이는 text 위젯에 옵션으로 yscrollcommand=scrollbar.set를 설정하여 해결할 수 있다.
```
scrollbar = tk.Scrollbar(window, command=text.yview)
text = tk.Text(window, yscrollcommand=scrollbar.set)
```
- 그러나 위와 같이 코드를 작성하는 경우 Text 위젯은 아직 완전히 생성되지 않은 Scrollbar를 참조하게 되고, 동시에 Scrollbar 역시 Text의 메서드를 참조해야 하므로 두 위젯이 서로를 동시에 참조하는 순환 참조 문제가 발생하여 에러가 난다.
- 따라서 아래와 같이 `.config(yscrollcommand=scrollbar.set)`로 별도 설정해주어야 한다.

````python
text.config(yscrollcommand=scrollbar.set)
# yscrollcommand : 세로스크롤이 생기면서 마우스휠로 텍스트 위젯 스크롤하면 스크롤바도 이동
text.pack()    #앞서 만들어놓았던  text.pack() 코드를 여기로 옮긴다.
````

### ✅ filedialog 박스
- "열기"메뉴를 클릭했다면 파일 다이얼로그 박스가 뜨면서 사용자가 열고자 하는 파일을 선택할 수 있어야 한다.
- 또한 "저장"메뉴를 클릭했을 때에도 파일 다이얼로그 박스가 뜨면서 사용자가 저장하고자 하는 파일명을 지정할 수 있어야 한다.
- tkinter.filedialog 모듈은 파일 열기/저장 대화상자를 제공하는 모듈이다.
- 사용자가 파일을 선택하거나 저장할 수 있도록 도와준다.
- 파일 다이얼로그 박스를 사용하려면 tkinter.filedialog 라이브러리를 import 시켜야 한다.

- filedialog 메서드는 다음과 같다.

    <font size=2>

    | 함수                   | 설명                                                  |
    | -------------------- | --------------------------------------------------- |
    | `askopenfilename()`  | 단일 파일을 열기 위한 파일 선택 대화상자. 선택된 \*\*파일 경로(문자열)\*\*를 반환 |
    | `askopenfilenames()` | **여러 개의 파일**을 열기 위한 대화상자. **리스트 형태의 파일 경로들**을 반환    |
    | `askopenfile()`      | 단일 파일을 열고 **파일 객체**로 반환                             |
    | `askopenfiles()`     | 여러 파일을 열고 **파일 객체들의 리스트**를 반환                       |
    | `asksaveasfilename()` | 저장할 파일의 \*\*경로(문자열)\*\*를 선택할 수 있는 대화상자 |
    | `asksaveasfile()`     | 저장할 파일을 **파일 객체**로 반환                  |
    | `askdirectory()` | 디렉토리를 선택하는 대화상자. \*\*선택된 폴더 경로(문자열)\*\*를 반환 |

    </font>

````python
import tkinter as tk
from tkinter import filedialog

window = tk.Tk()
window.withdraw()  # 윈도우 창을 숨긴다.

def load_diary():
    filepath = filedialog.askopenfilename(parent=window)
    if filepath:
        with open(filepath, 'r', encoding='utf-8') as file:
            lines = file.read()
            text.delete('1.0', tk.END) # 기존 내용 삭제, Tkinter의 Text 위젯에서 좌표는 "줄.열" 형태로 표현된다.
            text.insert('1.0', lines) # 1행 0열에 lines에 있는 내용을 삽입
            file.close()
load_diary()
````

- 파일을 열 때는 읽기, 쓰기, 추가 등 사용 목적을 지정해야 한다. 이때 지정하는 방식을 파일 모드라고 하며, 여러 가지 종류가 존재한다.

| 모드 | 의미 | 설명 |
|------|------|------|
| `r`  | 읽기(Read) | 기본 모드. 파일이 존재해야 하며, 읽기 전용으로 엶. |
| `r+` | 읽기/쓰기(Read & Write) | 파일을 읽고 쓰기 가능. 파일이 존재해야 함. |
| `w`  | 쓰기(Write) | 파일을 새로 작성. 기존 파일이 있으면 내용을 덮어씀. |
| `w+` | 읽기/쓰기(Write & Read) | 파일을 새로 작성하면서 읽고 쓰기 가능. 기존 내용은 모두 삭제됨. |
| `a`  | 추가(Append) | 파일이 존재하면 끝에 내용을 추가. 없으면 새로 생성. |
| `a+` | 추가/읽기(Append & Read) | 파일 끝에 내용 추가 가능 + 읽기 가능. 파일 없으면 생성. |
| `x`  | 배타적 생성(Create) | 파일이 존재하면 에러 발생. 없으면 새 파일 생성. |
| `x+` | 생성/읽기/쓰기(Create & Read & Write) | 파일이 없을 때만 생성하고 읽기/쓰기 가능. |

- 인코딩(encoding)은 사람이 이해하는 문자(예: "안녕하세요")를 컴퓨터가 저장하거나 전송할 수 있도록 숫자(바이트)로 변환하는 과정이다.
- 반대로, 디코딩(Decoding)은 저장된 숫자(바이트)를 다시 문자로 바꾸는 과정이다.
- 'utf-8'은 전 세계 대부분의 언어를 지원하는 범용 인코딩방법이다.

````python
import tkinter as tk
from tkinter import filedialog
from tkinter import messagebox

def save_diary():
    filepath = filedialog.asksaveasfilename(parent=window)
    if filepath:
        lines = text.get('1.0', tk.END+'-1c') # 마지막 new line 없음
        with open(filepath, 'w', encoding='utf-8') as file:
            file.write(lines)

def programexit():
    if messagebox.askokcancel("Quit", "종료하시겠습니까?"):
        window.destroy()

def about():
    label = messagebox.showinfo("About", "일기장 프로그램 ver.1.0")
````

### ✅ 캔버스 위젯
- 이번에는 캔버스를 만들고 그림을 그려보자.
- 캔버스(Canvas) 위젯은 직선, 원, 사각형, 이미지, 텍스트 등 그래픽 요소를 그릴 수 있는 공간이다.
- Tkinter에서 사용자 정의 그래픽 UI나 그림판, 게임 요소 등을 구현할 때 자주 사용된다.
- 캔버스는 `tk.Canvas()` 함수를 통해 만든다.
- 이때 캔버스의 크기를 설정할 수 있는데 width, height 단위에 혼동이 있을 수 있으니 주의하자.
- 캔버스의 단위는 픽셀이며, 텍스트위젯의 단위는 글자 수이다.

````python
canvas = tk.Canvas(window, width=500, height=300) 
canvas.pack()
````

- 캔버스 위젯은 다양한 키보드/마우스 이벤트를 처리할 수 있다.
- 마우스와 키보드 이벤트는 각각 다음과 같다.

    <font size=2>

    🖱️ 마우스 이벤트

    |--------------|-------------------------------|
    | `<Button-1>`    | 마우스 왼쪽 버튼 클릭         |
    | `<Button-2>`    | 마우스 가운데 버튼 클릭       |
    | `<Button-3>`    | 마우스 오른쪽 버튼 클릭       |
    | `<Double-1>`    | 마우스 왼쪽 더블 클릭         |
    | `<B1-Motion>`   | 마우스 왼쪽 버튼 누른 채 이동 |
    | `<Motion>`      | 마우스 이동                   |
    | `<Enter>`       | 마우스가 위젯 안으로 들어옴   |
    | `<Leave>`       | 마우스가 위젯 밖으로 나감     |

    ⌨️ 키보드 이벤트

- 또한 캔버스 위젯은 선, 사각형, 원, 호, 다각형 등과 같은 다양한 도형을 만들 수 있다.
- 캔버스 위젯 메서드는 다음과 같다.

    <font size=2>

    | 메서드                                    | 설명                     | 주요 옵션                                |
    | -------------------------------------- | ---------------------- | ------------------------------------ |
    | `create_line(x1, y1, x2, y2, ...)`     | 직선을 그림 (다중 좌표 가능)      | `fill`, `width`, `dash`, `arrow`     |
    | `create_rectangle(x1, y1, x2, y2)`     | 사각형을 그림                | `outline`, `fill`, `width`           |
    | `create_oval(x1, y1, x2, y2)`          | 주어진 사각형 범위 안에 타원/원을 그림 | `outline`, `fill`, `width`           |
    | `create_arc(x1, y1, x2, y2)`           | 타원의 호(arc)를 그림         | `start`, `extent`, `style`           |
    | `create_polygon(x1, y1, ..., xn, yn)`  | 여러 점을 연결한 다각형 그림       | `fill`, `outline`, `smooth`, `width` |
    | `create_text(x, y, text='내용')`         | 지정한 위치에 텍스트 출력         | `font`, `fill`, `anchor`             |
    | `create_image(x, y, image=PhotoImage)` | 이미지를 캔버스에 표시           | `anchor`, `state`                    |
    | `create_window(x, y, window=위젯)`       | 캔버스에 다른 위젯을 삽입         | 예: `Button`, `Entry` 등               |

    </font>

- 아래의 코드를 한줄씩 확인해보자.

````python
prev_x, prev_y = None, None

def paint(event):
    global prev_x, prev_y
    if prev_x is not None and prev_y is not None:
        canvas.create_line(prev_x, prev_y, event.x, event.y, width=2, fill='black', capstyle=tk.ROUND, smooth=True)
    prev_x, prev_y = event.x, event.y

def reset(event):
    global prev_x, prev_y
    prev_x, prev_y = None, None

canvas.bind("<B1-Motion>", paint)  # 마우스 왼쪽 버튼을 누르고 이동할 때
canvas.bind("<ButtonRelease-1>", reset)  # 마우스 왼쪽 버튼을 뗐을 때 이전 좌표 초기화
````

### ✅ 색상 선택하여 그리기와 지우기

- 이번에는 그림을 그리는 펜의 색을 변경해보자.
- `tkinter.colorchooser 모듈의 askcolor()` 함수는 색상 선택 대화상자를 나타낸다.
- 선택된 색을 이용하여 그림을 그릴 수 있도록 설정하자.
- 추후에 버튼을 여러개 추가할 것을 대비하여 프레임으로 묶어놓자.
- 아래 코드를 이해해보고 일기장에 추가해보자.

````python
from tkinter import colorchooser

def choose_color():
    color = colorchooser.askcolor(title="색상 선택") ## ((R, G, B), '#rrggbb') 형식의 튜플 반환
    myColor = color[1]

toolbar = tk.Frame(window)
toolbar.pack(side="top", fill="x")

btn = tk.Button(toolbar, text="색상 선택", command=choose_color)
btn.pack(pady=10)
````

- 여기서 잠깐, myColor는 지역변수이다. 지역변수는 함수안에서만 그 값을 사용할 수 있다.
- choose_color()의 변수 myColor에 선택된 색을 `canvas.create_line()`에서 적용하려면 변수 myColor는 전역변수여야 한다.
- 따라서 `global` 키워드로 myColor를 전역변수로 선언한다.
- 그런 후, 선택된 색이 적용되도록 canvas.create_line()의 fill옵션을 myColor로 설정한다.

````python
def choose_color():
    global myColor
    color = colorchooser.askcolor(title="색상 선택") ## ((R, G, B), '#rrggbb') 형식의 튜플 반환
    myColor = color[1]

# def paint(event) 함수 정의 부분에서 수정한다.
# 또한 전역변수 myColor를 'black'으로 초기값을 설정해야 한다.
canvas.create_line(prev_x, prev_y, event.x, event.y, width=2, fill=myColor, smooth=True)
````

### ✅ 그림과 텍스트 모두 저장하기
- 아쉽게도 우리 프로그램에서는 save_diary() 함수에서 text.get(...)로 Text 위젯 내용만 가져와서 파일에 쓰고 있기 때문에 텍스트 내용만 저장된다.
- 그림과 텍스트가 모두 저장되도록 확장해보자.
- 그림을 이미지 파일로 저장하고, 텍스트는 텍스트 파일로 따로 저장해보자. 물론 하나의 PDF로 저장할 수 있지만 텍스트 수정이 필요할 수 있으므로 각각 저장한다.

````python
from PIL import ImageGrab

def save_diary():
    date_str = date_picker.get() # yyyy-mm-dd 형태

    # Canvas의 화면 좌표 가져오기
    x = window.winfo_rootx() + canvas.winfo_x()
    y = window.winfo_rooty() + canvas.winfo_y()
    x1 = x + canvas.winfo_width()
    y1 = y + canvas.winfo_height()

    # 저장할 디렉토리 선택
    file_path = filedialog.asksaveasfilename(
        defaultextension=".png",
        filetypes=[("PNG Files", "*.png"), ("All Files", "*.*")],
        initialfile=date_str
    )

    # bbox 방식으로 캡처
    img = ImageGrab.grab(bbox=(x, y, x1, y1))

    # PNG로 안전하게 저장
    img.save(file_path, format='PNG') #절대경로 지정

    # 텍스트 저장
    txt_file= file_path[:-4]+".txt"
    with open(txt_file, "w", encoding="utf-8") as f:
        f.write(text.get("1.0", tk.END).strip())

    messagebox.showinfo("저장 완료", f"{date_str} 일기가 저장되었습니다.")
````

- 이번에는 이미지 파일과 텍스트를 불러와보자.

````python
import os

def load_diary():
    date_str = date_picker.get()

    file_path = filedialog.askopenfilename(
        defaultextension=".png",
        filetypes=[("PNG Files", "*.png"), ("All Files", "*.*")],
        initialfile=date_str
    )

    img_file = file_path
    txt_file = img_file[:-4]+".txt"

    # Canvas 비우기
    canvas.delete("all")

    # 그림 불러오기
    if os.path.exists(img_file):
        img = tk.PhotoImage(file=img_file)
        canvas.create_image(0, 0, anchor="nw", image=img) # n:위쪽 가운데 (north), w:왼쪽 가운데 (west)
        canvas.image = img  # 참조 유지

    # 글 불러오기
    if os.path.exists(txt_file):
        with open(txt_file, "r", encoding="utf-8") as f:
            text.delete("1.0", tk.END)
            text.insert("1.0", f.read())
````
