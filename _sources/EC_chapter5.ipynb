- pygame을 실행시킬 화면을 만들어보자.
- 아래의 코드를 실행해보면 통통볼 게임이라고 적힌 검정색 창이 뜰 것이다.

```python
import pygame

pygame.init()

# 조절 가능한 윈도우를 설정한다.
screen = pygame.display.set_mode((800, 600), pygame.RESIZABLE)
pygame.display.set_caption('통통볼 게임')

running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:   #게임 창의 'X' 버튼(닫기 버튼)**을 눌렀을 때 발생
            running = False

pygame.quit()
```

- **게임 루프**: 게임을 실행시키는 지속적인 사이클로, 이벤트 처리, 게임 상태 업데이트, 그래픽 렌더링 등을 담당한다.

1. 이벤트 처리: 키보드 입력, 마우스 클릭 및 기타 이벤트와 같은 사용자 입력을 지속적으로 확인하여 사용자와 실시간으로 상호작용한다.

2. 게임 로직 업데이트: 이벤트를 기반으로 게임 상태를 업데이트한다. 키보드 입력에 따라 캐릭터를 이동시키거나, 충돌 여부를 확인하거나, 점수를 갱신하는 작업 등을 수행한다.

3. 렌더링: 게임 상태가 업데이트된 후, 업데이트된 상태를 화면에 나타낸다. 여기에는 스프라이트, 배경, UI 요소를 그리는 작업이 포함될 수 있다.

```python
running = True
while running:
    # 1. 이벤트 처리
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    # 2. 게임 로직 업데이트


    # 3. 렌더링
    screen.fill((255, 255, 255))  # 배경 흰색
    pygame.display.flip()  # 화면에 전체 디스플레이 업데이트
```

## <div style="background-color:rgba(208, 205, 208, 1); padding: 10px; border-radius: 5px;"> <mark style="background-color: rgba(208, 205, 208, 1);">**이벤트 처리 및 화면 업데이트**</mark></div>

- 게임 루프 내에서 이벤트를 처리하고 화면을 업데이트하는 과정을 좀 더 자세히 살펴보자.
- 이벤트 처리는 게임 종료, 키 입력, 마우스 이동 등 다양한 이벤트를 처리하는 것을 의미한다.
- 키 이벤트는 사용자가 키보드와 상호작용할 때 사용된다.
- 마우스 이벤트는 움직임, 클릭, 스크롤과 같은 마우스 동작에 의해 발생한다.
- Pygame은 이벤트를 관리하기 위해 이벤트 큐(queue)를 사용한다.
- 큐는 자료 구조의 한 종류로 먼저 들어온 데이터가 먼저 나가는 구조이다. 즉, 선입선출(FIFO: First in First out)방법으로 먼저 들어온 이벤트가 먼저 처리된다.
- 큐는 아래 그림과 같이 후단(rear)으로만 데이터(여기서는 이벤트)가 들어오고, 전단(front)으로만 데이터가 나간다.

### 🎨 pygame.draw 함수 정리
<font size=2>

| 함수명                     | 설명                       | 주요 인자                                                    |
| ----------------------- | ------------------------ | -------------------------------------------------------- |
| `pygame.draw.line()`    | 선을 그린다.               | surface, color, start\_pos, end\_pos, width=1            |
| `pygame.draw.lines()`   | 여러 점을 연결한 선을 그린다.   | surface, color, closed, pointlist, width=1               |
| `pygame.draw.aaline()`  | 안티앨리어싱된 선을 그린다.         | surface, color, start\_pos, end\_pos, blend=1            |
| `pygame.draw.aalines()` | 여러 점을 연결한 안티앨리어싱 선       | surface, color, closed, pointlist, blend=1               |
| `pygame.draw.rect()`    | 사각형을 그린다               | surface, color, rect, width=0, border\_radius=0          |
| `pygame.draw.ellipse()` | 타원형을 그린다.(rect에 맞춰 그림) | surface, color, rect, width=0                            |
| `pygame.draw.circle()`  | 원을 그린다.                | surface, color, center, radius, width=0                  |
| `pygame.draw.arc()`     | 호(arc)를 그린다.            | surface, color, rect, start\_angle, stop\_angle, width=1 |
| `pygame.draw.polygon()` | 다각형을 그린다.               | surface, color, pointlist, width=0                       |

    - Surface: 도형을 그릴 대상 (보통 screen 사용)
    - Color: 도형의 색상 (RGB 튜플)
    - 안티앨리어싱된 선: 선을 픽셀 단위로 그리면 톱니 모양의 계단 현상이 생기지만, 안티앨리어싱을 적용하면 부드럽게 보이는 선을 그릴 수 있다.

### 🧩 공 이동 시키기    
- 아래는 공을 이동시키는 코드이다.
- 공의 좌,우,위,아래의 위치가 화면의 가장자리와 부딪히면 방향을 전환한다.
- 이 코드를 반복문의 화면 초기화 후 게임요소를 그리는 부분에 넣어준다.

```python
ball.x += ball_speed[0]
ball.y += ball_speed[1]

# 벽에 부딪히면 튕기기
if ball.left <= 0 or ball.right >= 800: # 화면의 좌우 끝에 부딪히면
    ball_speed[0] = -ball_speed[0]      # 방향 전환
if ball.top <= 0 or ball.bottom >= 600: # 화면의 위아래 끝에 부딪히면
    ball_speed[1] = -ball_speed[1]      # 방향 전환
```

### 🧩 공 이동 속도 조절
- 위 코드로 공을 이동시키면 엄청 빠른 속도로 움직일 것이다. 그 이유는 프레임 속도를 제한하지 않았기 때문이다.
- 프레임 속도란 영화나 동영상에서 1초 동안 재생되는 프레임(화면 이미지)의 수를 의미하며, 단위로는 fps(frames per second)를 사용한다.
- 프레임 속도를 제한하지 않으면 while running: 루프가 빠르게 계속 실행되고 있고 그로 인해, 공 또한 빠르게 이동한다.
- 프레임 속도는 pygame.time.Clock() 클래스의 tick(n)함수로 조절할 수 있다. n은 초당 프레임(FPS)수를 의미한다.
- 너무 느리거나 빠르면 프레임의 수를 조절하면 된다.

```python
clock = pygame.time.Clock()
clock.tick(60)
```

### 🧩 패들 만들고 마우스와 연동시키기
- 패들은 공만들기 했던 것처럼 `pygame.Rect()`를 이용하여 사각형 패들을 만들 수 있다.
- 패들의 색은 블루의 값만 255로 설정하여 파란색으로 만든다.
- 이 패들을 사용자가 마우스를 통해 이동시키기 위해 `pygame.mouse.get_pos()`을 통해 마우스 커서의 현재 (x, y) 좌표를 가져온다.
- 이때 패들은 좌우로만 움직일 수 있도록 하기 위해 x 좌표만 사용하고 y 좌표는 무시할 것이다.
- paddle.centerx = mouse_x: 패들의 중심 x 좌표를 마우스의 x 위치로 맞춘다. 즉, 마우스를 좌우로 움직이면 패들도 마우스를 따라 좌우로 이동하게 된다.
- paddle.clamp_ip(screen.get_rect()): paddle이 화면 바깥으로 나가지 않도록 제한한다.

```python
# 패들
paddle = pygame.Rect(270, 580, 100, 15)
paddle_color = (0, 0, 255) #RGB 색상값

# 마우스로 패들 이동
mouse_x, _ = pygame.mouse.get_pos()
paddle.centerx = mouse_x
paddle.clamp_ip(screen.get_rect())

# 그리기
pygame.draw.rect(screen, paddle_color, paddle)
```

### 🧩 게임 실패 넣기
- 패들이 공을 튕기지 못했을 때 즉, 공이 윈도우의 아래단을 터치하면 게임종료 메시지창을 띄우고 게임이 종료되도록 한다.
- 아쉽게도 Pygame은 messagebox를 직접 생성하는 기능을 제공하지 않는다.
- 그래서 이전에 배웠던 tkinter 모듈의 messagebox를 활용해보자.
- messagebox는 tkinter로 만든 윈도우에서 동작하므로 tk.Tk() 모듈의 윈도우를 생성해야 한다.
- 이때 tk 윈도우와 pygame 윈도우가 동시에 보이면 안되므로 tk 윈도우는 `window.withdraw()`를 통해 숨김처리한다.
- 이후 게임에 실패했을 때에는 윈도우도 종료시켜줘야 한다. 그렇지 않으면 보이지 않지만 프로그램은 여전히 실행 중이다.


```python
import tkinter as tk
import tkinter.messagebox  # 메시지창을 위한 모듈 추가

window = tk.Tk()
window.withdraw()  # 메인 윈도우 숨김

# 바닥에 닿으면 게임 오버
if ball.bottom >= 600:
    tkinter.messagebox.showinfo("Game Over", "공을 놓쳤어요!\n게임을 종료합니다.")
    running = False
    window.destroy()
```

### 🧩 점수 추가
- 점수를 추가하면 더 재미있는 게임을 만들 수 있다.
- 공이 패들에 닿을 때마다 점수가 올라가도록 하고, 이를 화면에 표시해보자.
- `pygame.font.SysFont()`를 통해 기본 폰트와 크기를 설정할 수 있다.
- 나타낼 텍스트는 `font.render()`를 사용하여 작성해준다.
- 게임화면에 텍스트를 나타낼 때에는 `윈도우객체.blit()`를 사용한다.

```python
# 점수
score = 0
font = pygame.font.SysFont(None, 40)  # 기본 폰트와 크기 설정

# 볼이 패들에 닿을 때 점수 증가 코드 추가
score += 1  # 점수 증가

score_text = font.render(f"Score: {score}", True, (0, 0, 0))  # 스크린이 검정이면 흰색으로 할것
screen.blit(score_text, (10, 10))  # 좌측 상단에 표시
```
