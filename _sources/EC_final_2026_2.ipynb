- 웹앱은 사용자가 직접 입력하거나 선택하여 결과를 확인할 수 있는 기능을 포함해야 하며, 단순한 정보 제공 페이지가 아닌 실제로 활용할 수 있는 프로그램으로 구현하시오.

### 참고사항
- 지연 제출 시 1일당 1점 감점합니다.
- 제출 관련 안내사항 및 요구사항을 준수하지 않은 경우 감점될 수 있습니다. 감점 사유에는 파일명 형식 미준수, 제출물 누락, 불필요한 파일 업로드, 기타 공지된 제출 기준 미준수 등이 포함되며, 이에 한정되지 않습니다.
- 프로젝트를 수행하기 위해 모든 공개된 자료(서적, 논문, 온라인 자료 등)를 활용할 수 있으며, 인용 시에는 참고문헌으로 명시해야 합니다.
- 생성형 AI도 활용 가능하며, 생성형 AI의 답변을 활용한 경우에는 질문과 답변을 어떻게 활용했는지 작성해야 합니다.
- 참고문헌과 생성형 AI 활용이 많은 것이 감점 사유가 되지 않으며, 활용을 하였으나 미 작성한 경우 감점이 됩니다.
