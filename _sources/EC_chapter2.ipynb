## <div style="background-color:rgba(208, 205, 208, 1); padding: 10px; border-radius: 5px;"> <mark style="background-color: rgba(208, 205, 208, 1);">GUI란?</mark></div>

- **GUI**(Graphical User Interface)는 **사용자가 컴퓨터나 프로그램을 쉽게 조작할 수 있도록 그래픽 요소를 활용한 인터페이스**를 의미한다.

- **GUI의 주요 특징**
  - 그래픽 요소 활용: 아이콘, 버튼, 창, 메뉴 등을 사용하여 직관적으로 조작 가능하다.
  - 마우스 및 터치 지원: 키보드뿐만 아니라 마우스, 터치스크린을 활용 가능하다.
  - 사용자 친화적: 명령어 입력 방식(CLI)보다 배우기 쉽고 접근성이 높다.
  - 멀티태스킹 지원: 여러 창을 동시에 띄워서 작업 가능하다.

- **GUI의 구성 요소**
  - 윈도우(Window): 프로그램이 실행되는 창
  - 아이콘(Icon): 프로그램이나 파일을 나타내는 작은 그림
  - 버튼(Button): 클릭하여 명령을 실행하는 요소
  - 메뉴(Menu): 사용 가능한 기능 목록을 제공하는 인터페이스
  - 스크롤바(Scrollbar): 화면을 넘겨서 추가 내용을 볼 수 있도록 함

- **GUI 주요 개발 도구**
  - Tkinter, PyQt 등으로 가능하나 우리 수업에서는 Tkinter를 사용한다.

- GUI는 사용자의 접근성과 사용 편의성을 높이기 위해 만들어졌으며, **이벤트 기반 프로그래밍** 방식으로 동작한다.
- 이벤트 기반 프로그래밍은 사용자 입력에 따라 프로그램이 즉시 반응하는 방식으로, 키보드 입력, 마우스 위치 감지, 클릭 등의 이벤트를 통해 사용자와 실시간으로 상호작용한다.
- 이러한 방식은 모바일 앱, 웹 애플리케이션 등 다양한 응용 프로그램에서 널리 활용된다.

## <div style="background-color:rgba(208, 205, 208, 1); padding: 10px; border-radius: 5px;"><mark style="background-color: rgba(208, 205, 208, 1);">샘플 소스 코드</mark></div>

- 아래 샘플 소스 코드를 실행시켜보자.
- 작은 윈도우 하나가 열리면서 <font color='green'>'Hello, Tkinter!'</font>라는 메시지가 보일 것이다. 
- GUI 기반 프로그램을 만들기 위해, 우리는 먼저 하나의 윈도우 창을 만들어 보았다.

````python
import tkinter as tk

window = tk.Tk()
window.title("Tkinter 테스트")
window.geometry("300x200")

label = tk.Label(window, text="Hello, Tkinter!")
label.pack()

window.mainloop()
````

### ✅ 모듈과 모듈 불러오기
**모듈**은 서로 관련있는 함수나 클래스를 모아 놓은 집합이다. 파이썬에는 random, math, sys, ... 등 수 많은 모듈들이 정의되어 있고 필요할 때 불러서 쓸 수 있다. 모듈은 한번만 불러오면 현재 소스코드 파일을 닫을 때까지 사용할 수 있으므로 여러 번 불러올 필요가 없다. 따라서 일반적으로 모듈을 불러오는 코드는 소스코드의 맨 위에 모아 놓는다.
모듈을 불러오는 키워드는 import이다.
모듈을 불러오는 방법은 두 가지가 있다.

- 첫 번째는 `import 모듈명`으로 불러온다. 
- 예를 들어, tkinter 모듈을 불러오려면 아래와 같이 쓰면 된다. 
- 이때 as 키워드를 통해서 tkinter의 별칭을 만들어 줄 수 있다.
- tkinter 모듈안에는 위젯을 만들 수 있는 많은 함수들이 있는데 `모듈명.함수명()`의 형식으로 점(.)으로 연결하여 사용한다.

````python
import tkinter as tk

window = tk.Tk()

tk.Label(window)
tk.Button(window)

window.mainloop()
````

- 두 번째는 `from 모듈명 import *`로 불러온다. 여기서의 *은 모든 것이라는 의미를 담고 있는데 즉, tkinter 모듈안에 모든 것을 불러온다는 뜻이다. 
- 이와 같이 불러온 경우에는 `모듈명.함수명()이 아니라 함수명()만` 써야한다.</font>
- from tkinter import *와 함께 다른 모듈(math, os 등)에서도 *를 쓰면 어떤 함수/클래스가 어디서 왔는지 알기 어렵기 때문에 첫 번째 방법을 권장한다.

````python
from tkinter import * 

window = Tk()
Label(window)
Button(window)

window.mainloop()
````

- 객체? 클래스? 인스턴스? 단어가 새롭고 어려울 것이다. 잠깐 그 개념을 익히고 가보자.
- **객체**는 데이터와 그 데이터에 관련되는 동작(절차, 방법, 기능)을 모두 포함한 개념이다.

- 예로, 자동차라는 객체는 메이커, 모델, 색상, 연식, 가격 등의 속성을 가지고 있고, 주행하기, 방향바꾸기, 주차하기, 음성인식 등의 동작을 한다.

- 클래스는 서로 관련된 데이터(속성, 값)와 해당 데이터를 다루는 함수(동작)들을 하나로 묶어 추상화한 것으로, 객체를 정의하기 위한 템플릿(폼, 포맷, 양식), 객체에 대한 설계도라고 볼 수 있다.

### ✅ 윈도우의 속성 설정
- 윈도우 인스턴스는 다양한 속성을 설정할 수 있는 여러 메서드를 제공한다.

<font size=2>

| 메소드 | 설명 | 예제 |
|--------|------|------|
| `title(text)` | 창 제목 설정 | `window.title("내 Tkinter 창")` |
| `geometry("WxH+X+Y")` | 창 크기 및 위치 설정 | `window.geometry("500x300+100+100")` |
| `resizable(width, height)` | 창 크기 조정 가능 여부 설정 | `window.resizable(False, False)` |
| `iconbitmap(file)` | 창 아이콘 변경 (Windows 전용) | `window.iconbitmap("icon.ico")` |
| `configure(bg=color)` | 창 배경색 변경 | `window.configure(bg="lightblue")` |
| `state(state)` | 창 상태 변경 (`normal`, `iconic`, `zoomed`) | `window.state("zoomed")` |
| `attributes(option, value)` | 창 투명도, 항상 위 설정 등 | `window.attributes("-alpha", 0.8)` |
| `update()` | 창을 강제로 업데이트 | `window.update()` |
| `destroy()` | 창을 닫고 종료 | `window.destroy()` |

</font>



- 아래는 윈도우 제목을 '온도 변환 프로그램', 윈도우 크기를 '350x180'으로 설정한 코드이다.

- 따라서 window.mainloop()을 호출하여 윈도우가 종료되지 않고, 사용자 이벤트(입력)를 계속해서 기다리는 상태로 유지한다.
- window.mainloop() 코드는 사용자가 윈도우를 닫을 때까지 계속해서 이벤트를 처리하라는 뜻이다.

### ✅ 레이블 위젯
- 레이블 위젯은 tkinter에서 텍스트나 이미지를 화면에 표시하는 위젯이다. 
- 버튼처럼 클릭 이벤트가 필요하지 않은 정적인 정보(제목, 설명, 아이콘 등)를 표시할 때 사용한다.
- 레이블 위젯의 옵션은 다음과 같다.

<font size=2>

| 옵션 이름         | 설명                                                             |
| ------------- | -------------------------------------------------------------- |
| `text`        | 라벨에 표시할 문자열                                                    |
| `font`        | 글꼴 설정 (예: `("Arial", 12, "bold")`)                             |
| `fg`          | 글자 색 (`foreground`의 줄임말, 예: `"red"`)                           |
| `bg`          | 배경 색 (`background`의 줄임말, 예: `"white"`)                         |
| `width`       | 라벨의 너비 (문자 기준)                                                 |
| `height`      | 라벨의 높이 (문자 기준)                                                 |
| `padx`        | 내부 좌우 여백 (padding x)                                           |
| `pady`        | 내부 상하 여백 (padding y)                                           |
| `relief`      | 테두리 스타일 (`flat`, `sunken`, `raised`, `groove`, `ridge`)        |
| `borderwidth` | 테두리 두께                                                         |
| `anchor`      | 텍스트 위치 (`n`, `s`, `e`, `w`, `center` 등)                        |
| `justify`     | 여러 줄일 때 텍스트 정렬 (`left`, `center`, `right`)                     |
| `image`       | 이미지 표시 (PhotoImage 객체 사용)                                      |
| `compound`    | 텍스트와 이미지 동시 표시 방식 (`top`, `bottom`, `left`, `right`, `center`) |

</font>

### ✅ 위젯 레이아웃 하기
- 위젯을 추가한 후에는 반드시 레이아웃 메서드를 통해 윈도우에 위젯이 배치되도록 해야 한다.
- GUI 레이아웃을 설정하는 메서드에는 grid(), pack(), place()가 있다.

<div align="center"><img src="https://haesunbyun.github.io/common/images/comps/image4.PNG" width=300>  </div>

- **grid()**:
  - 보이지 않는 격자 셀에 위젯을 배치함.
  - 특정 행과 열을 지정하여 배치할 수 있음
  - rowspan과 columnspan을 이용하여 여러 행과 여러 열에 걸쳐 배치할 수 있음
  - 가장 많이 사용됨 (정렬이 쉽고 유지보수하기 편리함)

- **pack()**:
  - 위젯을 위에서 아래로(세로 방향) 또는 왼쪽에서 오른쪽으로(가로 방향) 자동 배치
  - 간단한 UI에서는 유용하지만, 복잡한 UI에는 적합하지 않음

- **place()**:
  - 좌표(x, y)를 직접 지정하여 위젯을 원하는 위치에 배치
  - 픽셀 단위로 정확한 배치를 원할 때 사용하지만, 창 크기가 변경되면 깨질 수 있음

<div align="center"><img src="https://haesunbyun.github.io/common/images/comps/image3.png" width=300></div>

````python
# label을 윈도우에 배치한다.
label_1.grid(row=0, column=0, columnspan=2) 
label_2.grid(row=1, column=0) 
label_3.grid(row=2, column=0)
````

## <div style="background-color:rgba(208, 205, 208, 1); padding: 10px; border-radius: 5px;"><mark style="background-color: rgba(208, 205, 208, 1);">실행파일 만들기</mark></div>

- 온도변환 프로그램을 실행파일로 만들어보자.
1. **소스 코드 저장하기**
    - 하나의 .ipynb 또는 .py 파일에 하나의 프로그램에 대한 완성된 전체 소스코드를 넣어주기 위해 새 파일을 만든다.
    - 메뉴에서 [파일(File)] → [새 파일(New File)] → 파일이름: temperature.py -> 저장 경로 설정한 후 엔터
    - 전체 소스 코드를 temperature.py에 복사한다.
2. **.py → .exe로 변환**
    - PyInstaller 설치 (한 번만 설치하면 됨)
        - 메뉴에서 [Terminal] → [New Terminal]을 열어 pip install pyinstaller 입력 후 엔터
    - 아이콘 이미지를 넣어주는 경우(생략가능)
        - .ico 확장자의 아이콘 파일이 필요하다. (예: app_icon.ico) PNG나 JPG는 안 된다. .ico 확장자 변환 사이트를 활용하면 된다.
    - 실행파일 만들기
        - (아이콘을 넣지 않은 경우) 터미널에서 pyinstaller --onefile --noconsole temperature.py 엔터
        - (아이콘을 넣는 경우) pyinstaller --onefile --noconsole --icon=app_icon.ico temperature.py
        - 맥북의 경우 --onefile 옵션 대신에 --onedir 옵션으로 넣어준다.

    - 생성된 .exe 파일 위치
        - ./dist/temperature.exe

- 실행 파일을 만들었다면, 친구나 가족에게 전달하여 별도의 설치 과정 없이 PC에서 프로그램을 실행해볼 수 있다.
