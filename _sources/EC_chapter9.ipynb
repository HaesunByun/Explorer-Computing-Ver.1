### 웹 서비스를 구성하는 세 가지 기본 구성 요소

* 프런트엔드: 화면 구성을 처리
* 백엔드: 화면에 보여줄 정보를 처리
* API: 프런트엔드와 백엔드가 정보를 주고받을 수 있도록 일정한 규칙을 정하고, 그 규칙에 따라 동작하도록 설계된 인터페이스
  
- 이번 시간에는 Streamlit으로 웹 앱을 구현하고, GitHub + Streamlit Cloud를 통해 웹에서 바로 앱을 확인하고 실행해보자
    - GitHub: 코드와 파일을 저장하고 버전 관리를 하는 저장소 역할
    - Streamlit Cloud: GitHub에 있는 Streamlit 앱 코드를 가져와서 실제로 웹에서 실행할 수 있도록 호스팅해주는 서비스
        - 내부적으로 간단한 서버 역할은 하지만, 일반적인 백엔드 서버(데이터베이스 연동, API 처리 등)처럼 복잡한 서버 기능까지 제공하는 것은 아님

## <div style="background-color:rgba(208, 205, 208, 1); padding: 10px; border-radius: 5px;"><mark style="background-color: rgba(208, 205, 208, 1);">**Streamlit이란?**</mark></div>

- Streamlit은 파이썬 기반의 오픈소스 웹 애플리케이션 프레임워크로, 주로 데이터 시각화, 머신러닝 모델 시연, 대시보드 제작에 사용된다.
- Streamlit은 Python만으로 쉽고 빠르게 데이터 중심 웹 앱을 만들 수 있는 도구이다.

- 핵심 특징
    - 빠른 웹 앱 개발: Python 코드 몇 줄만으로 브라우저에서 동작하는 앱 제작 가능, HTML, CSS, JS를 따로 몰라도 됨
    - 실시간 인터랙티브: 슬라이더, 버튼, 체크박스 등 위젯으로 사용자 입력을 받아 실시간 데이터 업데이트 가능
    - 데이터 시각화와 친화적: Pandas, Matplotlib, Plotly등과 바로 연동 가능
    - 쉬운 배포: Streamlit Cloud, Heroku, Railway 등으로 손쉽게 웹에 배포 가능

- 브라우저가 열리면서 샘플 앱 화면이 나타나면, Streamlit이 정상적으로 설치된 것이다.

### 간단 예제 테스트 해보기
- 이제 streamlit 코드를 작성하여 웹에서 결과를 확인해보자.
- 아래 코드는 간단한 Streamlit 예제를 보여주는 코드이다.

````python
%%writefile streamlit_app.py
import streamlit as st
import pandas as pd
import numpy as np

st.title("간단한 Streamlit 예제")
st.write("데이터와 그래프를 보여주는 예제입니다.")

# 샘플 데이터
data = pd.DataFrame(
    np.random.randn(20, 3),
    columns=['A', 'B', 'C']
)

st.line_chart(data)
````

- 사용자 앱을 실행시키기 위해 새로운 터미널(메뉴 → New Terminal)을 열거나 혹은 기존 터미널의 서버 종료 후(Ctrl + C) 동일한 터미널에서 `streamlit run 파일명`을 통해 실행시킬 수 있다.
- 그러나 매번 실행 코드만 분리하여 파일로 만든후([파일(File)] → [새 파일(New File)]) 실행하기에는 좀 번거로운 일이다.

- 첫번째 라인 `%%writefile streamlit_app.py` 코드는 셀에 적힌 코드 내용을 파일로 저장하라는 명령이다. 실행하면 현재 폴더에 streamlit_app.py 파일이 생긴다.
- 이와 같이 코드를 작성해 놓으면 매번 실행파일을 별도로 만드는 작업을 하지 않아도 되므로 실습과 실행 관리를 쉽게 한다.

- 이제 `streamlit run streamlit_app.py`를 실행시켜보자.

## <div style="background-color:rgba(208, 205, 208, 1); padding: 10px; border-radius: 5px;"><mark style="background-color: rgba(208, 205, 208, 1);">**개발 환경 설정**</mark></div>

- 방금 Streamlit으로 만든 웹페이지는 로컬 컴퓨터의 브라우저에서 열린 것이므로 진정한 웹 서비스로 볼 수 없다.
- 이제 누구나 볼 수 있도록 웹에 배포해보자.
- Streamlit 앱을 웹에 배포하려면 몇 가지 요소가 필요하다.

1. 앱 코드(.py 파일)
    - Streamlit으로 작성된 코드

2. 깃허브(GitHub) 저장소
    - 코드를 온라인에 올려야 Streamlit Cloud에서 가져올 수 있음
    - 계정 필요, 새 리포지토리 생성

3. requirements.txt
    - 앱에서 사용하는 Python 라이브러리 목록
    - Streamlit Cloud가 앱 실행 환경을 구성할 때 필요

4. Streamlit Cloud 계정
    - 깃허브 리포지토리와 연결하여 앱 배포 가능
    - 무료/유료 플랜 있음

### 깃허브 회원가입 및 저장소 생성

#### 회원 가입
- 깃허브는 코드와 파일을 온라인에 저장하고 버전 관리 및 협업을 할 수 있는 플랫폼이다.
- 무료 계정과 유료 계정이 있으며, Streamlit 앱을 배포하는 목적이라면 대부분 무료 계정만으로 충분하다.
- 무료 계정은 다음과 같이 사용할 수 있다.
    - 공개(공용) 저장소: 무제한 생성 가능
    - 개인 저장소: 무료로 최대 제한 없이 생성 가능
    - 협업: 공개 저장소에서는 누구나 기여 가능, 개인 저장소에서는 초대된 사람만 접근 가능
    - 기본 기능: 버전 관리, 이슈 트래킹, Actions(자동화) 제한적 사용
- 유료 계정(GitHub Pro / Team / Enterprise)
    - 개인용 Pro: 개인 저장소와 협업 기능 강화, 더 많은 Actions 사용 가능
    - 팀/기업용: 팀 관리, 보안 기능, 대규모 저장소 관리, 전용 지원 제공    
- 깃허브에 가입하기 위하여 https://github.com/ 링크에 접속해보자.
- 화면 오른쪽에 보이는 Sign up 버튼을 클릭하여 회원가입을 진행한다.
- 이메일, 비밀번호, 사용자명 등을 입력한다. 이때 <font color="red"> 사용자명은 차후 생성될 깃허브 홈페이지 주소에 포함되므로 잘 생각하여 결정하기 바란다.</font>
    - 예) https://haesunbyun.github.io에서 *haesunbyun*
- 정보 입력이 끝나면 깃허브 가입에 사용한 이메일로 code가 전송된다.
- 여러분은 학생이니 'Apply for your GitHub Student Benefits'를 선택하여 더 많은 혜택을 얻기 바란다.

#### 저장소(repository) 생성

- 깃허브 가입을 완료했다면 이제는 코드를 올려놓기 위한 저장소를 생성해보자.
- 개인 깃허브 우측 상단에 보면 동그라미로 사용자가 누구인지 표시해주는 그림이 있다.
<div align="center"><img src="https://haesunbyun.github.io/common/images/comps/image25.png" width=300></div>

- 해당 그림을 클릭하여 select bar에서 두 번째에 위치한 Your repositories를 클릭한다. New 버튼을 찾아 새로운 저장소를 만든다.
- 저장소 이름은 <font color="red">Explorer-Computing-여러분의 이니셜</font>로 하고, Public으로 선택한다. 물론 원하는 저장소 이름으로 해도 된다.
- README 파일 추가 옵션은 체크한다. README 파일은 이 저장소를 방문했을 때 가장 먼저 보게 되는 파일로, 추후에 저장소 사용에 대한 소개, 개요 및 지침등을 작성해 놓을 수 있다.
- Create repository를 클릭하여 저장소를 만든다.

### 깃허브에 필요 파일 옮기기
- Streamlit Cloud가 필요한 패키지를 자동 설치해주기 위해서는 requirements.txt 파일안에 필요한 모듈들을 명시해 놓아야 한다.

### Streamlit 회원가입 및 깃허브 연동

#### 회원 가입
- Streamlit 앱을 웹으로 배포하기 위해서는 Streamlit 회원 가입을 해야 한다.
-  https://share.streamlit.io (Streamlit Community Cloud) 사이트로 접속하여 GitHub 계정으로 로그인한다.
- 처음 로그인 시 Streamlit Community Cloud 사용 권한을 요청하는 화면이 나오면 Authorize Streamlit 눌러서 이메일로 온 code로 인증한다.
<div align="center"><img src="https://haesunbyun.github.io/common/images/comps/image27.png" width=400></div>

#### 깃허브 연동

- 로그인이 되면, 깃허브에서 보았던 프로필 마크가 이 화면에서도 보일 것이다.
- 오른쪽 상단에 Create App을 눌러서 깃허브 저장소와 연동해보자.
    - Repository 선택: `username/저장소 이름`
    - Branch 선택: `main (또는 master)`
    - File path 입력: `streamlit_app.py` (처음 메인 화면으로 보일 파일 이름)
    - Deploy 버튼 클릭

<div align="center"><img src="https://haesunbyun.github.io/common/images/comps/image28.png" width=500></div>


- 잠시 빌드 후 → `https://여러분들의 저장소이름.streamlit.app` 주소가 생성된다.
- 이때, 만약 저장소 이름이 중복될 경우 뒤에 랜덤 문자열이 붙는다.
- 이제 이 링크를 통해 Streamlit 앱을 전 세계 누구에게나 공유 가능하다.
- 참고로, 깃허브 저장소를 Private로 변경해도 Streamlit 서비스를 지원 가능하다.
- 단, Streamlit 무료 플랜에서는 동시에 1개의 Private 저장소만 가질 수 있다.
- Private 저장소를 쓰려면 Streamlit이 GitHub의 저장소 권한(OAuth scope)을 받아야 한다.
- Private로 바꾸는 것은 어렵지 않으니 ChatGPT의 도움으로 변경해보기 바란다.
