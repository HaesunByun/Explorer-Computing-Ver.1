이번에 만들어볼 프로그램은 아래와 같은 인터페이스를 갖는 간식추천 프로그램이다.

<div align="center"><img src="https://haesunbyun.github.io/common/images/comps/image13.png" width=300>  </div>

- 아래 템플릿 소스 코드를 실행시키면 간단한 윈도우가 열릴 것이다.

```python
import tkinter as tk

# 창 설정
window = tk.Tk()
window.title("나만의 간식 추천기")
window.geometry("400x500")

# 이름 입력
label_name = tk.Label(window, text="이름을 입력하세요:")
label_name.pack(pady=5)
entry_name = tk.Entry(window)
entry_name.pack(pady=5)

# 취향 선택 (Radiobutton)
label_mood = tk.Label(window, text="어떤 간식을 좋아하나요?")
label_mood.pack(pady=5)

window.mainloop()
```

### ✅ 컨테이너 위젯

- Frame을 활용한 레이아웃 관리와 그룹화
  - Frame은 여러 위젯을 하나의 그룹으로 묶기 위한 컨테이너 위젯이다. 복잡한 GUI를 논리적으로 나누기 좋다.
  - Frame은 `tk.Frame()` 함수를 통해서 만든다.
  - 각각의 frame에는 pack(), grid(), place() 등 레이아웃을 독립적으로 적용할 수 있다.
  - 즉, 하나의 윈도우 안에 여러 개의 Frame을 만들고, 각 Frame 내부에서는 서로 다른 레이아웃 방식을 사용할 수 있기 때문에 복잡한 인터페이스도 유연하게 구성할 수 있다.
- 여기서는 라디오 버튼끼리, 체크박스끼리 각각 그룹화하기 위해 별도의 프레임을 사용한다.

```python
# 프레임을 만들어 부모 윈도우(window)에 넣어준다.
frame_mood = tk.Frame(window)
frame_mood.pack()
```

### ✅ 라디오 버튼 위젯
- 라디오 버튼(Radiobutton)은 여러 선택지 중 하나만 선택할 수 있게 해주는 위젯이다.
- `tk.Radiobutton()`를 통해서 만든다.
- 같은 그룹의 라디오 버튼들은 공유하는 변수를 통해 선택된 값을 식별한다.
- 아래 예에서는 변수 snack_type이 라디오 버튼의 공유변수이며, 선택된 value가 저장된다.
- 라디오 버튼들은 앞서 만든 frame에 넣어줌으로서 그룹화한다.

<div align="center"><img src="https://haesunbyun.github.io/common/images/comps/image14-1.png" width=300>  </div>

```python
# 간식 취향
snack_type = tk.StringVar()
snack_type.set("달콤한")  # 기본값 설정

#라디오 버튼을 만든다.
tk.Radiobutton(frame_mood, text="달콤한 간식", value="달콤한", variable=snack_type).pack(side="left", padx=5)
tk.Radiobutton(frame_mood, text="짭짤한 간식", value="짭짤한", variable=snack_type).pack(side="left", padx=5)
#건강한 간식 라디오 버튼은 직접 코드를 작성하세요.
```

<font size=2>

| `variable`           | 그룹을 구성할 공통 변수 (`tk.IntVar`, `tk.StringVar` 등)                    |
| `value`              | 이 버튼이 선택될 때 변수에 설정할 값                                            |
| `command`            | 선택 시 실행할 함수                                                      |
| `state`              | 버튼 상태 (`"normal"`, `"disabled"`)                                 |
| `bg` 또는 `background` | 배경색                                                              |
| `fg` 또는 `foreground` | 글자색                                                              |

| `indicatoron`        | 체크 표시 대신 버튼처럼 보이게 할지 여부 (`True`/`False`)                         |
| `selectcolor`        | 선택되었을 때 점 내부 색상                                                  |
| `activebackground`   | 마우스가 올라갔을 때 배경색                                                  |
| `activeforeground`   | 마우스가 올라갔을 때 글자색                                                  |
| `relief`             | 테두리 모양 (`"flat"`, `"raised"`, `"sunken"`, `"groove"`, `"ridge"`) |

### ✅ 체크박스 위젯

- 체크박스는 여러 항목 중 복수 선택이 가능할 때 사용한다.

<font size=2>

| `onvalue`           | 체크되었을 때 `variable`에 저장될 값 (기본값: 1)                    |
| `offvalue`          | 체크 해제되었을 때 `variable`에 저장될 값 (기본값: 0)                 |
| `command`           | 체크 상태가 변경될 때 실행할 함수                                   |

| `font`              | 텍스트의 폰트 설정                                            |
| `fg` / `foreground` | 텍스트 색상                                                |

| `anchor`            | 텍스트 정렬 (`"w"`, `"e"`, `"center"` 등)                   |
| `justify`           | 여러 줄 텍스트 정렬 방식 (`"left"`, `"center"`, `"right"`)      |
| `width`             | 체크박스의 너비 (문자 단위)                                      |

### ✅ 리스트박스 위젯

- 리스트박스(Listbox)는 여러 항목 중에서 하나 또는 여러 개의 항목을 선택할 수 있는 위젯이다.
- 항목은 insert() 메서드로 추가하며, 선택한 값은 curselection()이나 get()으로 가져올 수 있다.
    - curselection(): 선택된 항목의 **인덱스**(위치)를 튜플 형태로 반환, 예: (0, 2) → 0번과 2번 항목이 선택됨
    - get(index): 주어진 인덱스에 해당하는 실제 **값**(문자열)을 반환

<div align="center"><img src="https://haesunbyun.github.io/common/images/comps/image14-3.png" width=300> </div>    

| 옵션                  | 설명                                                                                                                                                    |
| ------------------- | ----------------------------------------------------------------------------------------------------------------------------------------------------- |
| `bg`                | 배경색 (background) 지정                                                                                                                                   |
| `fg`                | 글자색 (foreground) 지정                                                                                                                                   |
| `font`              | 글꼴과 크기 설정 (예: `("Arial", 12)`)                                                                                                                        |
| `height`            | 표시할 항목의 개수 (스크롤 전 기준)                                                                                                                                 |
| `width`             | 리스트박스의 문자 단위 폭                                                                                                                                        |
| `listvariable`      | `tkinter.StringVar`와 연결하여 항목 관리 가능                                                                                                                    |
| `selectmode`        | 선택 모드 설정 <br>• `browse`: 한 항목만 선택 (기본값)<br>• `single`: 클릭으로 하나만 선택<br>• `multiple`: 여러 개 선택 (클릭마다 toggle)<br>• `extended`: 여러 개 선택 (Shift, Ctrl 키 지원) |
| `exportselection`   | 다른 위젯과 선택 내용을 공유할지 여부 (기본 `True`)                                                                                                                     |
| `justify`           | 텍스트 정렬 방식 (`left`, `center`, `right`)                                                                                                                 |
| `relief`            | 테두리 모양 (`flat`, `sunken`, `raised`, `groove`, `ridge`)                                                                                                |
| `selectbackground`  | 선택된 항목의 배경색                                                                                                                                           |
| `selectforeground`  | 선택된 항목의 글자색                                                                                                                                           |
| `selectborderwidth` | 선택 영역의 테두리 두께                                                                                                                                         |
| `activestyle`       | 활성 항목 표시 스타일 (`underline`, `dotbox`, `none`)                                                                                                          |
| `xscrollcommand`    | 수평 스크롤바 연결                                                                                                                                            |
| `yscrollcommand`    | 수직 스크롤바 연결                                                                                                                                            |
| `state`             | `normal` 또는 `disabled` (편집/비활성화 여부)                                                                                                                   |
| `cursor`            | 마우스 커서 모양 변경 (예: `"hand2"`)                                                                                                                           |


```python
# 리스트박스 레이블 추가



# 리스트박스를 만든다.
listbox_snacks = tk.Listbox(window, selectmode="multiple", height=6)
snack_items = ["초콜릿", "감자칩", "견과류", "요거트", "쿠키", "과일"]
for item in snack_items:
    listbox_snacks.insert(tk.END, item)
listbox_snacks.pack(pady=5)    
```

## <div style="background-color:rgba(208, 205, 208, 1); padding: 10px; border-radius: 5px;"><mark style="background-color: rgba(208, 205, 208, 1);">**메시지 박스**</mark></div>

- 메시지 박스(message box)는 프로그램에서 사용자에게 정보를 전달하거나, 확인을 요청하거나, 경고/오류를 알릴 때 사용하는 팝업 창을 말한다.

### ✅ 메시지 박스 띄우기

- 메시지 박스를 띄우기 위해서는 먼저 messagebox 모듈을 불러와야 한다.
- messagebox 모듈은 tkinter에 들어있다.
- from tkinter import *를 쓰더라도 messagebox는 자동으로 안 들어온다. 반드시 `from tkinter import messagebox`가 필요하다.
- 메시지 박스의 종류와 예시는 다음과 같다.

<font size=2>

| 함수명                              | 용도                              | 예시                                                     |
| -------------------------------- | ------------------------------- | ------------------------------------------------------ |
| `showinfo(title, message)`       | 정보 메시지                          | `messagebox.showinfo("안내", "정상적으로 처리되었습니다.")`          |
| `showwarning(title, message)`    | 경고 메시지                          | `messagebox.showwarning("경고", "입력이 누락되었습니다.")`         |
| `showerror(title, message)`      | 에러 메시지                          | `messagebox.showerror("오류", "파일을 찾을 수 없습니다.")`         |
| `askquestion(title, message)`    | 예/아니오 질문 (결과: `'yes'`, `'no'`)  | `response = messagebox.askquestion("확인", "삭제하시겠습니까?")` |

```python
listbox_snacks.bind("<<ListboxSelect>>", on_select_snack)
```

### ✅ 간식 이미지 업데이트 함수 정의하기  

```python
def show_snack_image(snack_name):
    if snack_name in snack_images:
        img_label.config(image=snack_images[snack_name])
        img_label.image = snack_images[snack_name]  # 이미지가 계속 보이도록 설정하는 코드
    else:
        img_label.config(image='')  # 선택한 이미지가 없으면 초기화

def on_select_snack(event):
    """Listbox에서 간식을 선택하면 자동으로 이미지 변경"""
    if not listbox_snacks.curselection():
        show_snack_image(None) # 선택을 해제했을 때 이미지를 사라지게 하기 위함
        return
    # 복수 선택 가능이므로 첫 번째 선택만 사용
    idx = listbox_snacks.curselection()[0]
    snack_name = listbox_snacks.get(idx)
    show_snack_image(snack_name)
```
