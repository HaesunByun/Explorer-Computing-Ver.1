## <div style="background-color:rgba(208, 205, 208, 1); padding: 10px; border-radius: 5px;"> <mark style="background-color: rgba(208, 205, 208, 1);">**템플릿 소스 코드**</mark></div>

<div align="center"><img src="https://haesunbyun.github.io/common/images/comps/image18.png?v=1234" width=400>  </div>

- 캐릭터 이미지는 제 딸이 직접 그린 이미지입니다.😄
- 이번에 만들어볼 게임은 풍선이 랜덤한 위치에 랜덤한 속도로 여러개가 내려오고 어린 여자 아이가 풍선을 잡을 때마다 점수가 올라가는 게임이다.

````python
import pygame

# 초기화
pygame.init()
screen = pygame.display.set_mode((800, 600))
pygame.display.set_caption("풍선을 잡아라!")
clock = pygame.time.Clock()

score = 0
font = pygame.font.SysFont(None, 40)  # 기본 폰트와 크기 설정

# 메인 루프
running = True
while running:
    screen.fill((255, 255, 255))  # 배경 흰색

    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False  

    score_text = font.render(f"Score: {score}", True, (0, 0, 0))  # 검정색
    screen.blit(score_text, (10, 10))  # 좌측 상단에 표시
    
    pygame.display.flip()
    clock.tick(60)

pygame.quit()
````

### ✅ 객체, 클래스, 인스턴스
- **객체**는 데이터와 그 데이터에 관련되는 동작(절차, 방법, 기능)을 모두 포함한 개념이다.
- 이 세상에 존재하는 모든 것, 예를 들어, 학생, 자동차, 책상, 버튼 등 모든 것이 객체가 될 수 있으며, 모든 객체는 속성(attribute)과 동작(action, method)을 가지고 있다.
- 예로, 자동차라는 객체는 메이커, 모델, 색상, 연식, 가격 등의 속성을 가지고 있고, 주행하기, 방향바꾸기, 주차하기, 음성인식 등의 동작을 한다.
- 여기서 정의하고자 하는 게임 캐릭터 또한 하나의 객체이며, 캐릭터는 모양, 속성, 동작 등을 가진다.

<div align="center"><img src="https://haesunbyun.github.io/common/images/adt2.png" style="width:300px;"></div>

- 파이썬에서는 `클래스를 통해 객체를 구현`한다.
- 클래스는 서로 관련된 데이터(속성, 값)와 해당 데이터를 다루는 함수(동작)들을 하나로 묶어 추상화한 것으로, 객체를 정의하기 위한 템플릿(폼, 포맷, 양식), 객체에 대한 설계도라고 볼 수 있다.

### ✅ 클래스 정의

- 클래스를 정의할 때에는 `class`라는 키워드로 시작한다.

<div style="border:2px solid #ccc; border-radius:8px; padding:10px;">

```python
class 클래스명:
     클래스 변수
     인스턴스 메서드(method)
```
</div>

- `클래스 변수`는 해당 클래스의 모든 인스턴스가 공통(공유)으로 값을 가지는 변수를 말한다. 일반적으로 클래스 내부, 인스턴스 메서드(함수)의 바깥쪽에 정의한다.

- `인스턴스 메서드`는 클래스의 함수를 의미한다. 첫 번째 인자는 항상 self를 가진다.

- 실 세계의 객체인 개(Dog)를 생각해 보도록 하자.

- '개'라는 객체가 공통적으로 가지는 **동작**(operations)은 꼬리를 흔들고, 네 다리로 뛰며, 짖기도 한다.
- 우리는 '개'라는 객체를 개념적으로 위와 같이 **추상화**를 할 수 있다.

- 그럼 '개'라는 객체를 소프트웨어적으로 실체를 만들기 위해 클래스를 통해 템플릿을 먼저 만들어보자.

````python
class Dog:
  pass
````

#### 클래스 변수 정의
- 클래스안에 변수 attr1과 attr2를 만들어보자.  attr1과 attr2는 클래스 변수이다. 클래스 변수는 동일 클래스에서 만들어지는 모든 인스턴스가 공통적으로 갖는 값이다.
- Dog 클래스를 통해 만들어지는 모든 인스턴스는 'mammal'(포유류)과 'dog'(개)라는 값(value)을 갖는다.

````python
class Dog:
    attr1 = "mammal"
    attr2 = "dog"
````

#### 인스턴스 메서드 정의
- 클래스 Dog에 **인스턴스 메서드**를 만들어보자.
- 인스턴스 메서드는 클래스의 함수이므로 함수를 정의하는 것과 동일하게 만든다.
<div style="border:2px solid #ccc; border-radius:8px; padding:10px;">

```python
def show(self):
    인스턴스 메서드 본문
```
</div>

- 여기서 show()메서드는 첫 번째 인자로 self를 갖는다. self는 인스턴스 그 자신을 의미한다. 
- 동일 클래스의 인스턴스를 여러 개 만들 수 있으므로 각 인스턴스 자신을 가르키는 self가 필요하다. 아직 self가 이해되지 않더라도 실망하지 말자. 대학교 2학년 때 쯤 C++로 객체지향프로그래밍을 처음으로 배우기 시작했을 때 self를 이해하는데 한참 걸렸던것 같다.ㅜㅜ
- 메서드안에서 클래스변수에 접근하려면 self.클래스변수명으로 접근한다.
- 클래스를 정의했으면 파이썬이 인식하도록 아래 셀을 실행시켜준다.

````python
class Dog:
    '''
    Dog Class by haesun
    '''
    attr1 = "mammal"
    attr2 = "dog"

    def show(self):
        print("I'm a", self.attr1)
        print("I'm a", self.attr2)
````

- 예를 들어, Dog() 클래스의 인스턴스 momo를 만드려면 momo = Dog()와 같이 쓴다. 이를 통해 Dog라는 객체가 인스턴스화 된다.

- 글로벌 영역에서 클래스안에 정의된 클래스 변수에 접근할 때에는 다음과 같이 점(.)으로 연결한다.
<div style="border:2px solid #ccc; border-radius:8px; padding:10px;">

#### 인스턴스 메서드 호출
- 클래스안에 정의된 인스턴스 메서드를 호출하려면 다음과 같이 쓴다.
<div style="border:2px solid #ccc; border-radius:8px; padding:10px;">

```python
인스턴스명.메서드명()
클래스명.메서드명(인스턴스명)
```
</div>
    
- 예를 들어, Dog 객체는 show()라는 인스턴스 메서드를 갖고 있는데 show() 메서드를 호출하려면 momo.show()와 같이 쓴다.
- 이는 또한 여러분들이 listVar = list()를 만들어 놓고 listVar.append(item) 했던 것과 같은 원리이다.
- 인스턴스명이 아니라 클래스명으로도 직접 호출할 수 있다. 예를 들어, Dog.show(momo)와 같이 호출할 수도 있다. 이 때에는 show()괄호안에 인스턴스명을 명시해줘야 한다.

````python
momo.show()
Dog.show(momo)
````

#### 인스턴스 변수
- 인스턴스 변수는 각 인스턴스마다 값을 가지는 변수를 말한다.
- 메서드에서 정의되며 `self.변수명`으로 접근한다.
- 예를 들어, 아래 인스턴스 메서드 run()에서 self.speed는 인스턴스 변수이다.

````python
class Dog:
    '''
    Dog Class by haesun
    '''
    attr1 = "mammal"
    attr2 = "dog"

    def show(self):
        print("I'm a", self.attr1)
        print("I'm a", self.attr2)
        print(f"My speed is {self.speed}")

    def run(self, speed):
        self.speed = speed # 인스턴스변수
        self.show() #class안에서 호출
        sound = 'walwal'  # 지역변수
````

- 인스턴스 변수는 동일 클래스 내 정의된 메서드 어디에서든지 `self.변수명`으로 활용 가능하다.
- 예를 들어,  self.speed 변수는 run() 메서드에서 만들어졌지만 show() 메서드에서 self.speed 값에 접근할 수 있다.
- 그래서 지역변수라고 하지 않고 인스턴스 변수라고 한다.

````python
momo = Dog()
momo.run(10)
momo.show()
````

- 인스턴스 변수를 클래스 밖에서 접근하려면 다음과 같이 쓴다.
<div style="border:2px solid #ccc; border-radius:8px; padding:10px;">

```python
인스턴스명.인스턴스변수
```
</div>

#### 상속과 스페셜 메서드

- 상속은 기존 클래스를 바탕으로 새로운 클래스를 정의하는 방법으로, 부모 클래스(상위 클래스)의 속성과 메서드를 자식 클래스(하위 클래스)가 물려받아 사용할 수 있게 한다.
- 클래스를 정의할 때, 상위 클래스를 상속받으면 그 안에 이미 정의된 여러 메서드들도 함께 사용할 수 있다.
- class Dog:처럼 아무것도 명시적으로 상속하지 않으면, 자동으로 object 클래스를 상속받는다.
이처럼 상속받은 요소들 중에는 다양한 메서드(method) 가 포함되어 있으며, 그중 이름 앞뒤에 이중 언더바(__)가 붙은 메서드들을 스페셜 메서드(special method) 또는 매직 메서드(magic method) 라고 한다. 즉, 이미 특정 기능으로 이름 붙여진 메서드이다.
- 이와 같은 스페셜 메서드들을 여러분들이 만든 클래스에서 동일 이름으로 재정의하면 클래스와 관련된 특정 기능을 수행 시킬 수 있다. 이 가운데 몇 개의 스페셜 메서드를 활용해보자.
- `__init__`(self, ...) : 생성자라고도 불리우며, 인스턴스가 생성 될때 자동으로 실행된다.
- `__del__`(self) : 소멸자라고도 불리우며, 인스턴스 삭제 시 자동으로 실행된다.
- `__call__` : 인스턴스명( )으로 호출했을 때 실행된다.
- `__str__`(self) : 객체를 문자열로 표현할 때 사용되며, print() 함수 호출 시 자동으로 사용된다.
- `__repr__`(self): 객체의 공식적인 문자열 표현을 위해 사용되며, 개발자가 이해하기 쉬운 형태로 객체를 표현할 때 사용
- `__len__`(self): 컨테이너의 길이를 반환할 때 사용 (예: len() 함수 호출 시)
- `__getitem__`(self, key): 컨테이너의 특정 항목을 가져올 때 사용 (예: obj[key])
- `__setitem__`(self, key, value): 컨테이너에 특정 항목을 설정할 때 사용. (예: obj[key] = value)
- `__delitem__`(self, key): 컨테이너의 특정 항목을 삭제할 때 사용 (예: del obj[key]).
- `__iter__`(self): 컨테이너를 반복할 때 사용 (예: 반복문에서 사용).
- `__next__`(self): 반복자의 다음 항목을 반환할 때 사용
- `__call__`(self, ...): 객체를 함수처럼 호출할 수 있게 함 (예: obj()).
- `__eq__`(self, other): 두 객체의 동등성(equality) 비교를 위해 사용.(예: obj1 == obj2).
- `__ne__`(self, other): 두 객체의 비동등성(non-equality) 비교를 위해 사용
(예: obj1 != obj2).
- `__lt__`(self, other), `__le__`(self, other), `__gt__`(self, other), `__ge__`(self, other):
객체 간의 크기 비교를 위해 사용 (각각 <, <=, >, >=).

#### 클래스의 초기화
- Dog 클래스의 `__init__`메서드는 self를 포함하여 4개의 매개변수를 갖고 있다.
- 따라서 인스턴스를 만들때 전달인자를 3개 줘야 한다. 4개라 아니라 3개이다. self는 인스턴스 그 자신이기 때문이다.
- 예를 들어, momo를 만들 때 전달인자로 "poodle", "brown","bowwow"를 넣어주자. 이때 momo는 self, "poodle"은 breed, "brown"은 color, "bowwow"는 sound로 매핑된다.

    ```python
    momo = Dog("poodle", "brown","bowwow")
    ```

````python
class Dog:
    '''
    Dog Class
    '''
    attr1 = "mammal"
    attr2 = "dog"

    def __init__(self, breed, color, sound):
        self.breed = breed  #견종
        self.color = color  #색깔
        self.sound = sound  #소리
        self.speed = 0

    def show(self):
        print("I'm a", self.attr1)
        print("I'm a", self.attr2)
        print(f"I'm a {self.breed}")
        print(f"My color is {self.color}")
        print(f"My sound is {self.sound}")
        print(f"My speed is {self.speed}")

    def run(self, speed):
        self.speed = speed

momo = Dog("poodle", "brown","bowwow")
momo.show()        
````

### ✅ 플레이어 이동

- 이동은 스프라이트의 기본적인 요소로, 게임 세계와 상호작용할 수 있게 해준다. 
- Player 클래스에 update() 메서드도 정의해보자.
- update() 메서드는 사용자 입력을 바탕으로 스프라이트의 위치를 업데이트한다.
- pygame.key.get_pressed()는 현재 키가 눌려 있는 상태를 매 프레임마다 확인하는 함수이다.
- 반면, 지난 시간에 배웠던 for문과 pygame.event.get()을 이용한 if event.type == pygame.KEYDOWN: 방식은, 키를 계속 누르고 있어도 이벤트가 한 번만 발생한다는 특징이 있다.
- 키보드의 상하좌우 키에 따라 self.rect 즉 이미지의 위치를 업데이트한다.

````python
def update(self):
    keys = pygame.key.get_pressed()
    if keys[pygame.K_LEFT]:
        self.rect.x -= 5
    if keys[pygame.K_RIGHT]:
        self.rect.x += 5
    if keys[pygame.K_UP]:
        self.rect.y -= 5
    if keys[pygame.K_DOWN]:
        self.rect.y += 5
````

### 🚀**해보기 4: 캐릭터 이동**

- 앞서 정의한 Player 클래스의 update() 메서드의 예시처럼 키보드 입력을 기반으로 캐릭터의 위치를 조정하여 이동시킬 수 있다.
- 키보드의 입력을 기반으로 캐릭터가 상하좌우로 이동하였다.
- 더 복잡한 이동을 구현하려면 물리 기반 계산을 적용할 수 있다. 
- 물리 기반 이동은 self.velocity = pygame.math.Vector2(0, 0)를 통해 속도/중력/점프/관성 등을 포함시킬 수 있다. 
- self.velocity는 플레이어의 이동 속도/방향을 저장하는 벡터이다. 
- 아래 코드는 중력을 적용하고, 캐릭터의 이동도 좌,우,점프만 가능하도록 작성한 것이다.
- 또한 캐릭터가 화면 밖으로 나가는 것을 방지하도록 좌우 이동 제한을 걸어보자.

````python
class Player(pygame.sprite.Sprite):
    def __init__(self):
        super().__init__()
        self.image = pygame.image.load('./images/player.jpg').convert_alpha()
        self.image = pygame.transform.scale(self.image, (80, 100))
        self.rect = self.image.get_rect()
        self.rect.center = (400, 300)
        self.velocity = pygame.math.Vector2(0, 0)  #속도/중력/점프/관성 등을 포함

    def update(self):
        self.velocity.x = 0  # 매 프레임 좌우 속도 초기화
        self.velocity.y += 0.5 #

        keys = pygame.key.get_pressed()
        if keys[pygame.K_LEFT]:
            self.velocity.x = -5  
        if keys[pygame.K_RIGHT]:
            self.velocity.x = 5
        if keys[pygame.K_UP] and self.rect.bottom >= 600: # 윗쪽 방향키 + 지면에 있을때만 점프
            self.velocity.y = -10  # Jump

        self.rect.x += self.velocity.x
        self.rect.y += self.velocity.y

        # 아래 화면으로 벗어나는 것을 방지   
        if self.rect.bottom >= 600:
            self.rect.bottom = 600
            self.velocity.y = 0
            
        # 좌우 화면으로 벗어나는 것을 방지    
        if self.rect.left <= 0:
            self.rect.left = 0
        if self.rect.right >= 800:
            self.rect.right = 800            
````

### ✅ 목표물 행동 정의하기

- 만들어볼 목표물은 풍선이 아래로 내려오는 상황을 표현하려고 한다. 
- 플레이어는 목표물을 잡으면 된다. 

- 아래 코드는 풍선이 아래로 내려오는 상황을 표현하고 있다.
- `pygame.transform.scale(original_image, (width, height))`는 이미지의 사이즈를 조정하고 있다.
- `self.rect.center = (random.randint(50, 750), random.randint(-100, -40))`는 풍선이 만들어지는 위치이다. 
- 이때, y의 값이 음수인데 이는 보이지 않는 화면 위쪽에서 만들어져서 아래로 자연스럽게 내려오는 듯하게 보이기 위함이다.
- `self.velocity = random.uniform(1.0, 2.5)`는 풍선 낙하 속도이다. 
- `pygame.mask.from_surface()` 픽셀 단위 충돌 감지를 위해 마스크(mask)를 만드는 함수이다.
    - 기본 충돌 판정은 rect(사각형) 영역으로 이루어지는데, 이미지가 둥글거나 투명한 부분이 많으면 사각형 판정은 부정확할 수 있다.
    - 예를 들어, 풍선 이미지가 동그랗고 주변이 투명하다면, 사각형 충돌은 풍선 밖 투명 영역도 포함해서 오작동할 수 있다.
    - 이미지에서 투명하지 않은(불투명한) 픽셀만 마스크로 표시한다.
    - 두 객체의 마스크가 겹치는 부분이 있을 때만 충돌로 인식한다.

````python
import random

class Enemy(pygame.sprite.Sprite):
    def __init__(self):
        super().__init__()
        self.image = pygame.image.load('./images/balloon.jpg').convert_alpha()
        self.image = pygame.transform.scale(self.image, (60, 80))
        self.rect = self.image.get_rect()
        self.rect.center = (random.randint(50, 750), random.randint(-100, -40))  # 화면 보이지 않는 위쪽에서 만들어져서 아래로 내려오게 함
        self.velocity = random.uniform(1.0, 2.5)  # 풍선 낙하 속도
````

### ✅ 목표물 스프라이트 그룹화
- `pygame.sprite.Group()`을 통해 이번에는 변수 balloon_group을 만들고 이 그룹안에 풍선들을 넣으려고 한다.
- 처음에는 풍선 3개를 스프라이트 그룹안에 넣기 위해 `balloon_group.add(Enemy())`로 Enemy() 클래스 객체를 추가한다.
- while 반복문 안에서는 balloon_group의 개수가 3보다 적을 때 Enemy() 객체를 추가한다.
- `update()와 draw()`를 통해 그룹 내 모든 스프라이트를 동시에 업데이트하고, 그려준다.

````python
balloon_group = pygame.sprite.Group() 

# 처음에 3개 풍선
for _ in range(3):
    balloon_group.add(Enemy())


# 풍선이 너무 적으면 자동 추가하는 코드를 반복문안에 넣어준다.
while len(balloon_group) < 3:
    balloon_group.add(Enemy())    
````

### ✅ 목표물 이동
- 풍선은 중력을 적용받으며 내려와야 하고, 화면 밑으로 사라지면 제거해야 한다.
- 이 동작을 Enemy() 클래스에 추가해준다. 
- 또한 update()와 draw(screen) 함수 호출은 반복문 안에 넣어준다.

````python
def update(self):
    self.rect.y += self.velocity
    if self.rect.top > 600:  # 화면 아래로 나가면 제거
        self.kill()

balloon_group.update()        
balloon_group.draw(screen)
````

이를 통해 클래스의 개념을 이해하고 객체지향 프로그래밍의 기초를 직접 경험할 수 있었다.
또한 게임 클래스를 정의하여 캐릭터의 위치와 속도를 관리하면서 상태와 동작을 체계적으로 다루는 방법을 배웠다.
스프라이트 기능을 활용하여 화면 요소들을 구조적으로 관리하는 경험도 쌓았다.
이 과정을 통해 실제로 움직이는 캐릭터를 만들며 성취감을 느끼고, 향후 더 복잡한 게임 개발에도 확장 가능한 코드를 작성할 수 있는 자신감을 얻을 수 있게 될 것이다.
