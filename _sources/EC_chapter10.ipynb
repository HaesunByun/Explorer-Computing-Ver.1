### 텍스트 표시 관련 메서드

- streamlit은 텍스트를 화면에 출력하기 위해 다양한 방법을 제공한다.

<font size=2>

| 메서드              | 설명                      | 예시                                             |
| ---------------- | ----------------------- | ---------------------------------------------- |
| `st.text()`      | 단순 텍스트 출력 (스타일 없음)      | `st.text("Hello Streamlit")`                   |
| `st.markdown()`  | 마크다운 문법 지원 텍스트 출력       | `st.markdown("**굵게** *기울임*")`                  |
| `st.write()`     | 자동 형식 인식(문자열, 숫자, 객체 등) | `st.write("문자열", 123, df)`                     |
| `st.title()`     | 페이지 제목 스타일 출력 (큰 글씨)    | `st.title("대제목")`                              |
| `st.header()`    | 섹션 제목 스타일 출력 (중간 크기)    | `st.header("중제목")`                             |
| `st.subheader()` | 작은 제목 스타일 출력            | `st.subheader("소제목")`                          |
| `st.caption()`   | 캡션(작은 설명글) 출력           | `st.caption("설명 텍스트")`                         |
| `st.code()`      | 코드 블록 출력 (문법 강조 가능)     | `st.code("print('Hello')", language="python")` |
| `st.latex()`     | LaTeX 문법 수식 출력          | `st.latex(r"E = mc^2")`                        |

</font>

```python
%%writefile Ex_ch10_text.py
import streamlit as st
import pandas as pd

# 단순한 텍스트를 표시
st.text("Hello, Streamlit!")

# Markdown 문법을 지원하여 헤더, 볼드, 이탤릭, 하이퍼링크 등 다양한 텍스트 스타일 가능
st.markdown("# 헤더 1")
st.markdown("## 헤더 2")
st.markdown("**굵은 텍스트**")
st.markdown("_기울임 텍스트_")
st.markdown("[링크로 이동](https://streamlit.io)")

# write()함수는 다목적 함수로 텍스트, 마크다운, 데이터, 객체 등 다양한 요소를 자동으로 감지하여 인식함
st.write("# 제목") 
st.write("Hello, Streamlit!")
st.write("**굵은 텍스트**와 *기울임 텍스트*를 함께 사용할 수 있습니다.")
st.write(12345)                        # 숫자
st.write({"이름": "홍길동", "나이": 23})   # 딕셔너리
st.write(pd.DataFrame({"A": [1,2,3]})) # 데이터프레임

st.title("대제목")
st.header("중제목")
st.subheader("소제목")
st.caption("설명 텍스트")
st.code("print('Hello')", language="python")
st.latex(r"E = mc^2")
```

### 데이터 표시 관련 메서드
- streamlit은 데이터를 테이블, 배열, 또는 데이터 프레임 형태로 쉽게 표시할 수 있는 다양한 방법을 제공한다.
- 이를 활용하면 데이터 시각화와 탐색이 간단해진다.

<font size=2>

| 메서드              | 설명                             | 예시                                                  |
| ---------------- | ------------------------------ | --------------------------------------------------- |
| `st.table()`     | 정적 테이블로 표시 (단순 출력)             | `st.table(df.head())`                               |
| `st.dataframe()` | 데이터프레임을 인터랙티브하게 표시 (스크롤/정렬 가능) | `st.dataframe(df)`                                  |
| `st.json()`      | JSON 데이터 구조 보기 좋게 표시           | `st.json({"name": "Alice", "age": 25})`             |
| `st.metric()`    | 지표(숫자+증감률) 표시                  | `st.metric(label="매출", value="1000", delta="+10%")` |

</font>

```python
%%writefile Ex_ch10_data.py
import streamlit as st
import pandas as pd
import numpy as np

# 샘플 데이터
data = {"교과목": ["컴퓨팅기초", "컴퓨팅탐색", "컴퓨팅핵심"], "강좌수": [19, 1, 4]}
list_data = [["컴퓨팅기초", 19], ["컴퓨팅탐색", 1], ["컴퓨팅핵심", 4]]
df = pd.DataFrame(data)

st.write("### 간단히 표시")
st.write(df)

# st.table(): 정적인 테이블을 표시. 간단한 데이터 프레임이나 배열을 깔끔하게 표시할 때 유용.
st.write("### 데이터프레임 정적 테이블")
st.table(df)

st.write("### 리스트 정적 테이블")
st.table(list_data)

# st.dataframe(): 대화형 데이터 테이블을 표시. 정렬, 검색 등 기본적인 테이블 조작 및 너비와 높이 조정 가능.
st.write("### 데이터프레임 대화형 테이블")
st.dataframe(df, width=500, height=200)

# Numpy 배열
array = np.random.rand(5, 3)
st.write("### Numpy 배열")
st.dataframe(array)

# st.json(): JSON 데이터 구조를 트리 형태로 표시. 펼치기, 접기가 가능
# JSON (JavaScript Object Notation), 자바스크립트 객체 표기법
# 사람이 읽고 쓰기 쉽고, 컴퓨터가 분석하기도 쉬운 텍스트 기반 데이터 형식
json_data = {"Name": "Alice", "Age": 25, "Hobbies": ["Reading", "Traveling"]}
st.write("### JSON 데이터")
st.json(json_data)

#지표(숫자+증감률) 표시 
st.metric(label="매출", value="1000", delta="+10%")

# HTML을 직접 사용하여 커스텀 테이블을 표시
# HTML로 테이블 생성
html_table = """
<table style="border: 1px solid black;">
  <tr>
    <th>Name</th>
    <th>Age</th>
  </tr>
  <tr>
    <td>Alice</td>
    <td>25</td>
  </tr>
  <tr>
    <td>Bob</td>
    <td>30</td>
  </tr>
</table>
"""
st.write("### HTML 테이블")
st.markdown(html_table, unsafe_allow_html=True)
```

### 레이아웃과 컨테이너

- streamlit에서는 앱의 **레이아웃**(layout)과 **컨테이너**(container)를 통해 UI 구성 요소들을 정리하고, 동적인 배치 및 스타일링을 쉽게 조정할 수 있다.

<font size=2>

| 메서드             | 설명                     | 예시                                   |
| --------------- | ---------------------- | ------------------------------------ |
| `st.sidebar`    | 사이드바 영역에 UI 배치         | `st.sidebar.button("사이드 버튼")`        |

#### 사이드바(Sidebar)
- 사이드바를 통해 주요 내비게이션 요소를 별도의 영역에 배치할 수 있다.
- 사이드바는 화면 왼쪽에 고정된 공간을 사용하여 앱의 전반적인 구조를 더욱 직관적으로 만들 수 있다.
- 아래는 streamlit의 사이드바를 활용한 웹페이지의 예이다.

```python
%%writefile Ex_ch10_sidebar1.py
import streamlit as st

# 사이드바 생성
st.sidebar.header("Sidebar Menu")

st.title("컴퓨팅 탐색")

# 라디오 버튼 메뉴
menu = st.sidebar.radio("메뉴 선택", ["홈", "분석", "설정"])
st.write(f"선택한 메뉴: {menu}")

# 셀렉트 박스 메뉴
option = st.sidebar.selectbox("Select an option", ["Option 1", "Option 2", "Option 3"])
st.write(f"You selected {option}")

# 체크박스
agree = st.sidebar.checkbox("동의합니다")
if agree:
    st.write("체크박스를 선택했습니다!")

# 슬라이더: .slider("문장", 최솟값, 최댓값, 초기값)
value = st.sidebar.slider("값을 선택하세요", 0, 100, 50)
st.write(f"선택한 값: {value}")

# 텍스트 입력
name = st.sidebar.text_input("이름 입력", "홍길동")
st.write(f"입력한 이름: {name}")
```

사이드바 안에 라디오박스, 셀렉트박스, 체크박스, 슬라이더, 텍스트 기반의 메뉴를 구성한 예를 보여준다.

```python
%%writefile Ex_ch10_sidebar2.py
import streamlit as st

st.title("📖 책 목차 예제")

# 목차 구성
toc = {
    "1장. 들어가기": [
        "1.1 소개",
        "1.2 준비물"
    ],
    "2장. 본문": [
        "2.1 데이터 불러오기",
        "2.2 분석하기",
        "2.3 시각화"
    ],
    "3장. 마무리": [
        "3.1 요약",
        "3.2 참고자료"
    ]
}

# 상위 챕터 선택
chapter = st.sidebar.selectbox("📂 챕터 선택", list(toc.keys()))

# 하위 섹션 선택
section = st.sidebar.radio("📑 섹션 선택", toc[chapter])

# 본문 출력
st.header(section)

if section == "1.1 소개":
    st.write("이 책은 Streamlit 학습을 위한 예제집입니다.")
elif section == "2.2 분석하기":
    st.write("여기서는 데이터 분석 방법을 다룹니다.")
# ... 필요에 따라 내용 추가
```

- 위의 코드는 하위 섹션의 수가 많아지면 if~elif문이 길어진다.

```python
%%writefile Ex_ch10_sidebar3.py
import streamlit as st

st.title("📚 책 목차 예제 ")

# 목차 정의
toc = {
    "1장. 들어가기": ["1.1 소개", "1.2 준비물"],
    "2장. 본문": ["2.1 데이터 불러오기", "2.2 분석하기", "2.3 시각화"],
}

# ----------------------------
# 섹션별 내용 함수
# ----------------------------
def section_intro():
    st.subheader("1.1 소개")
    st.write("""
    이 책은 Streamlit을 학습하기 위한 예제집입니다.  
    주요 목표는 데이터 시각화와 대시보드 제작을 빠르게 익히는 것입니다.
    """)

def section_setup():
    st.subheader("1.2 준비물")
    st.write("""
    Python 3.8 이상과 Streamlit이 설치되어 있어야 합니다.  
    설치는 `pip install streamlit` 으로 가능합니다.
    """)

def section_load():
    st.subheader("2.1 데이터 불러오기")
    st.write("""
    Pandas를 이용한 데이터 분석하기 위해 데이터를 불러옵니다.
    """)

def section_analysis():
    st.subheader("2.2 분석하기")
    st.write("""
    Pandas를 이용한 데이터 분석 예제를 다룹니다.
    """)

def section_visual():
    st.subheader("2.3 시각화")
    st.write("""
    matplot, seaborn를 이용한 데이터를 시각화 합니다.
    """)
# ----------------------------
# 목차 매핑
# ----------------------------
toc_function = {
    "1.1 소개": section_intro,
    "1.2 준비물": section_setup,
    "2.1 데이터 불러오기": section_load,
    "2.2 분석하기": section_analysis,
    "2.3 시각화": section_visual,    
}

# 선택 저장용 변수 초기값
selected = "1.1 소개"

st.sidebar.header("📖 목차")

for chapter, sections in toc.items():
    st.sidebar.markdown(f"### {chapter}")  # 챕터 제목
    for sec in sections:
        if st.sidebar.button(sec, key=sec):  # 각 섹션을 버튼으로 표시
            selected = sec


# ----------------------------
# 본문 표시
# ----------------------------
toc_function[selected]()   # 선택된 함수 실행
```

#### 컬럼(Column)
- 컬럼을 사용하여 UI 요소들을 단을 나누어 배치할 수 있다
- 각 컬럼은 1부터 12까지의 비율을 설정하여 크기를 나눌 수 있다.

```python
%%writefile Ex_ch10_columns.py
import streamlit as st

# 두 개의 컬럼을 생성
#col1, col2 = st.columns(2)

col1, col2 = st.columns([2, 1])  # 첫 번째 컬럼은 두 배 크기

with col1:
    st.header("Column 1")
    st.write("This is the first column.")
    st.text_area("col1", height=300)

with col2:
    st.header("Column 2")
    st.write("This is the second column.")
    st.text_area("col2", height=300)
```

#### 탭(Tab)
- 탭을 사용하면 여러 개의 탭 페이지를 만들어 사용자에게 선택할 수 있는 인터페이스를 제공한다.

```python
%%writefile Ex_ch10_tabs.py
import streamlit as st

# 탭 생성
tab1, tab2, tab3 = st.tabs(["Tab 1", "Tab 2", "Tab 3"])

with tab1:
    st.write("This is tab 1")

with tab2:
    st.write("This is tab 2")

with tab3:
    st.write("This is tab 3")
```

#### 컨테이너(Container)
- 컨테이너는 여러 위젯이나 요소들을 하나의 그룹으로 묶어준다.
- 반복적으로 같은 레이아웃을 적용하거나, 특정 부분을 별도로 다루고 싶을 때 유용하다.

```python
%%writefile Ex_ch10_container.py
import streamlit as st

# 컨테이너를 사용하여 그룹화
with st.container():
    st.header("컨테이너 안에 헤더")
    st.write("컨테이너 안에 텍스트")

# 3개의 컨테이너를 미리 만들고 변수에 저장
containers = [st.container() for _ in range(3)]

# 각각 다른 내용 출력
for i, container in enumerate(containers):
    with container:
        st.write(f"컨테이너 {i+1}")

# 두 번째 컨테이너만 업데이트
if st.button("2번째 블록 갱신"):
    with containers[1]:
        st.write("🟢 2번째 컨테이너 내용이 갱신됨!")
```

참고
- enumerate(): 리스트나 튜플을 반복하면서, 동시에 “인덱스 번호”와 “값”을 한꺼번에 꺼내주는 함수

```python
fruits = ["사과", "배", "바나나"]

for i, fruit in enumerate(fruits):
    print(i, fruit)
```

#### Expander (접을 수 있는 영역)
- expander는 사용자가 필요할 때 펼쳐서 내용을 볼 수 있도록 하는 접을 수 있는 영역이다.
- 이 기능은 긴 내용을 숨기거나, 옵션을 설정할 때 유용하다.

```python
%%writefile Ex_ch10_expander.py
import streamlit as st

# Expander 영역
with st.expander("더 많은 정보를 보려면 클릭"):
    st.write("This is the hidden content that appears when expanded.")

    # 클릭 가능한 버튼
    if st.button("버튼 클릭"):
        st.success("버튼이 눌렸어요!")

    # 체크박스
    agree = st.checkbox("이용 약관에 동의합니다.")
    if agree:
        st.info("동의해주셔서 감사합니다 🙌")

    # 드롭다운(리스트박스)
    option = st.selectbox("테마를 선택하세요", ["라이트", "다크", "시스템 기본"])
    st.write(f"선택한 테마: {option}")

    # 슬라이더
    volume = st.slider("볼륨 설정", 0, 100, 50)
    st.write(f"현재 볼륨: {volume}")
```

### 차트 관련 메서드
- streamlit에서 차트 메서드는 데이터를 시각적으로 표현하기 위한 다양한 방법을 제공한다.
- 차트 메서드는 내장 함수와 외부 라이브러리와의 연동을 통해 손쉽게 추가할 수 있다.     
    - 간단한 차트: st.line_chart, st.bar_chart 등
    - 복잡한 커스텀 차트: Matplotlib, Seaborn
    - 대화형 차트: Plotly, Bokeh
    - 지도 기반 시각화: Pydeck

<font size=2>

| 메서드                    | 설명                                 | 예시                            |
| ---------------------- | ---------------------------------- | ----------------------------- |
| `st.line_chart()`      | 선형 차트(line chart)           | `st.line_chart(df)`           |
| `st.area_chart()`      | 영역 차트(area chart)           | `st.area_chart(df)`           |
| `st.bar_chart()`       | 막대 차트(bar chart)            | `st.bar_chart(df)`            |
| `st.scatter_chart()`   | 산점도(scatter plot)           | `st.scatter_chart(df)`        |
| `st.map()`             | 위도/경도 데이터가 있을 때 지도에 점을 표시      | `st.map(df)`                  |
| `st.pydeck_chart()`    | PyDeck 기반의 인터랙티브 지도 시각화    | `st.pydeck_chart(deck_obj)`   |
| `st.pyplot()`          | Matplotlib 차트를 표시   | `st.pyplot(fig)`              |

</font>

```python
%%writefile Ex_ch10_chart1.py
import streamlit as st
import pandas as pd
import numpy as np

data = pd.DataFrame(
    np.random.randn(20, 3),
    columns=['A', 'B', 'C']
)
st.line_chart(data)
st.area_chart(data)
st.bar_chart(data)
```

- PyDeck은 Streamlit에서 지도 기반 시각화를 쉽게 구현할 수 있도록 도와주는 파이썬용 데이터 시각화 라이브러리이다.
- 위도(latitude)와 경도(longitude) 데이터를 이용하여 지도 위에 표현할 수 있다.

```python
%%writefile Ex_ch10_chart2.py
import streamlit as st
import pandas as pd
import pydeck as pdk

# 위도, 경도 데이터 준비
data = pd.DataFrame({
    'city': ['Seoul', 'Daejeon', 'Gwangju', 'Mokpo', 'Busan', 'Daegu'],
    'lat': [37.5665, 36.3504, 35.1595, 34.8110, 35.1796, 35.8714],
    'lon': [126.9780, 127.3845, 126.8526, 126.3925, 129.0756, 128.6014],
    'population': [9700000, 1500000, 1500000, 240000, 3400000, 2400000]    
})

st.write("📍 한국 주요 도시 위치")
st.map(data) 

# 3. Pydeck Chart
st.subheader("PyDeck Chart (st.pydeck_chart)")
layer = pdk.Layer(
    "ScatterplotLayer",  #점(원형 마커)을 표시
    data=data,
    get_position="[lon, lat]",
    get_radius=5000,  #점(원의 반지름)을 미터 단위로 지정. 5000이면 지상에서 약 5km 반경의 원으로 보임
    get_color=[255, 0, 0],
    pickable=True  #마우스로 점을 클릭했을 때 툴팁(정보창) 을 표시
)

# 툴팁 설정
tooltip = {
    "text": "도시: {city}\n인구: {population}",
    "style": {"color": "white"}
}

# 대한민국 전체가 보이는 위치로 설정
view_state = pdk.ViewState(
    latitude=36.5,  #지도의 중심 위도(lat), 36.5는 한반도의 대략적인 중앙 위도
    longitude=127.8, #지도의 중심 경도(lon), 127.8은 한국의 대략적인 중앙 경도
    zoom=6,
    pitch=0  # 지도를 기울이는 각도 0은 정면
)

st.pydeck_chart(pdk.Deck(layers=[layer], initial_view_state=view_state, tooltip=tooltip))
```

### 주요 위젯 메서드

<font size=2>

| 메서드                  | 설명             | 예시                                    |
| -------------------- | -------------- | ------------------------------------- |
| `st.button()`        | 버튼 생성          | `st.button("클릭")`                     |
| `st.checkbox()`      | 체크박스 생성        | `st.checkbox("동의")`                   |
| `st.radio()`         | 라디오 버튼 생성      | `st.radio("선택", ["A", "B"])`          |
| `st.selectbox()`     | 드롭다운 선택 박스     | `st.selectbox("옵션", ["1", "2", "3"])` |
| `st.slider()`        | 슬라이더(숫자 범위 선택) | `st.slider("범위", 0, 100, 50)`         |
| `st.text_input()`    | 텍스트 입력 상자      | `st.text_input("이름 입력")`              |
| `st.number_input()`  | 숫자 입력 상자       | `st.number_input("나이 입력")`            |
| `st.date_input()`    | 날짜 입력          | `st.date_input("날짜 선택")`              |
| `st.file_uploader()` | 파일 업로드         | `st.file_uploader("파일 선택")`           |

</font>

```python
%%writefile Ex_ch10_basic1.py

import streamlit as st
import pandas as pd

data = {"교과목": ["컴퓨팅기초", "컴퓨팅탐색", "컴퓨팅핵심"], "강좌수": [19, 1, 4]}
df = pd.DataFrame(data)

# 데이터 다운로드 버튼 추가
@st.cache_data
def convert_df_to_csv(df):
    return df.to_csv(index=False).encode('utf-8')

csv = convert_df_to_csv(df)

st.download_button(
    label="📥 데이터 다운로드",
    data=csv,
    file_name="data.csv",
    mime="text/csv",  #파일의 형식(콘텐츠 타입)을 브라우저에 알려주는 역할
)

# 버튼 추가
if st.button("Click me!"):
    st.write("Button clicked!")

# 체크박스 추가
agree = st.checkbox("I agree")
if agree:
    st.write("Thank you for agreeing!")

# 라디오 버튼 여러 선택지 중 하나를 선택
option = st.radio(
    "Choose your favorite color:",
    ["Red", "Blue", "Green"]
)
st.write("You selected:", option)

# 드롭다운 메뉴에서 하나의 선택지를 선택
option = st.selectbox(
    "Select your profession:",
    ["Engineer", "Artist", "Doctor"]
)
st.write("You selected:", option)

# 여러 선택지를 한 번에 선택
options = st.multiselect(
    "Select your hobbies:",
    ["Reading", "Traveling", "Gaming", "Cooking"]
)
st.write("You selected:", options)

# 숫자 범위를 조정할 수 있는 슬라이더
age = st.slider("Select your age:", 0, 100, 25)
st.write("Your age is:", age)

# 숫자를 직접 입력하거나 화살표로 조정
number = st.number_input("Enter a number:", min_value=0, max_value=100, value=50)
st.write("Your number is:", number)

```

```python
%%writefile Ex_ch10_basic2.py

import streamlit as st
import pandas as pd


# 한 줄의 텍스트를 입력
name = st.text_input("Enter your name:")
st.write("Hello,", name)

# 여러 줄의 텍스트를 입력
message = st.text_area("Enter your message:")
st.write("Your message is:", message)

# 날짜 선택
date = st.date_input("Pick a date:")
st.write("Selected date:", date)

# 시간 선택
time = st.time_input("Pick a time:")
st.write("Selected time:", time)

# 파일 업로드
uploaded_file = st.file_uploader("Choose a file:")
if uploaded_file is not None:
    st.write("File uploaded:", uploaded_file.name)

# 색 선택
color = st.color_picker("Pick a color:", "#00f900")
st.write("Selected color:", color)

# 여러 위젯을 그룹화하여 일괄 처리
with st.form("my_form"):
    name = st.text_input("Enter your name")
    age = st.number_input("Enter your age")
    submitted = st.form_submit_button("Submit")
    if submitted:
        st.write("Name:", name, "Age:", age)
```

### 미디어 요소
- 이미지, 오디오, 비디오 등 다양한 멀티미디어 콘텐츠를 앱에 추가한다.

<font size=2>

| 메서드          | 설명        | 예시                                      |
| ------------ | --------- | --------------------------------------- |
| `st.image()` | 이미지 표시    | `st.image("img.png", caption="예시 이미지")` |
| `st.audio()` | 오디오 파일 재생 | `st.audio("sound.mp3")`                 |
| `st.video()` | 비디오 파일 재생 | `st.video("video.mp4")`                 |

</font>

```python
%%writefile Ex_ch10_media.py
import streamlit as st
import numpy as np

# 이미지 파일 경로 사용, use_column_width는 이미지 크기를 자동으로 조정할지 여부
st.image("./images/balloon.jpg", caption="Example Image")

# URL에서 불러오기
st.image("https://haesunbyun.github.io/common/images/comps/player.jpg", caption="Image from URL")

# Numpy 배열 사용 (예: Matplotlib로 생성된 이미지)
# (100, 100, 3)는 (행, 열, 채널)
img = np.random.randint(0, 255, (100, 100, 3), dtype=np.uint8)
st.image(img, caption="Random Image")

# 오디오 파일 경로 사용
# https://haesunbyun.github.io/common/images/background.mp3
st.audio("./sound/background.mp3", format="audio/mp3")

# URL에서 오디오 재생
st.audio("https://haesunbyun.github.io/common/images/background.mp3", format="audio/mp3")

# 비디오 파일 URL에서 비디오 재생
st.video("https://haesunbyun.github.io/common/data/walking.avi")

# 사용자가 업로드한 파일을 미디어로 표시
# 이미지 업로드 후 표시
uploaded_image = st.file_uploader("Upload an image", type=["png", "jpg", "jpeg"])
if uploaded_image is not None:
    st.image(uploaded_image, caption="Uploaded Image")

# 오디오 업로드 후 재생
uploaded_audio = st.file_uploader("Upload an audio file", type=["mp3", "wav"])
if uploaded_audio is not None:
    st.audio(uploaded_audio)

# 비디오 업로드 후 재생
uploaded_video = st.file_uploader("Upload a video file", type=["mp4", "mov"])
if uploaded_video is not None:
    st.video(uploaded_video)

# Markdown으로 유튜브 등 외부 동영상 삽입
# streamlit 자체적으로 외부 동영상을 삽입하는 기능은 없지만, Markdown을 활용하면 YouTube 등 외부 비디오를 삽입 가능.
st.markdown(
    """
    <iframe width="560" height="315" src="https://www.youtube.com/embed/XKqCO_Er75Y"
    frameborder="0" allow="accelerometer; autoplay; clipboard-write; encrypted-media; gyroscope; picture-in-picture"
    allowfullscreen></iframe>
    """,
    unsafe_allow_html=True
)
```
