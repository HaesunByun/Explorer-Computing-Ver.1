## <div style="background-color:rgba(208, 205, 208, 1); padding: 10px; border-radius: 5px;"><mark style="background-color: rgba(208, 205, 208, 1);">Visual Studio Code(VS Code)</mark></div>

- VS code는 마이크로소프트에서 개발한 **무료 오픈소스 코드 편집기**이다. 가볍지만 강력한 기능을 갖추고 있어 많은 개발자들이 사용하고 있다.

장점은 다음과 같다.<br>
- **가볍고 빠름**: 무거운 IDE(IntelliJ, Eclipse 등)보다 훨씬 가볍고 빠르게 실행된다.
- **강력한 확장성**: 다양한 확장(Extensions)을 통해 기능을 쉽게 추가할 수 있다. (예: Python Jupyter, Git 등)
- **다양한 언어 지원**: 기본적으로 많은 언어를 지원하고, 플러그인을 통해 거의 모든 프로그래밍 언어를 사용할 수 있다.
- **Git 통합**: Git이 내장되어 있어 별도 도구 없이도 버전 관리가 가능하다.
- **디버깅 도구 내장**: Node.js, Python 등 다양한 언어에 대한 디버깅을 VS Code 내에서 실행할 수 있어 개발 생산성이 높다.
- **멀티 플랫폼 지원**: Windows, macOS, Linux 모두 지원한다.

단점은 아래와 같다.<br>
- **초기 설정 필요**: 언어별 환경 세팅을 직접 설치하고 설정해야 해서 초보자에게는 약간 번거로울 수 있다.
- **리소스 사용 증가**: 여러 확장 기능을 많이 설치하면 속도가 느려지고 메모리 사용량이 증가할 수 있다.
- **GUI 기반 작업엔 불편**: GUI 앱을 설계할 때 드래그앤드롭 방식의 UI 빌더가 지원되지 않아 효율적이지 않다.

## <div style="background-color:rgba(208, 205, 208, 1); padding: 10px; border-radius: 5px;"><mark style="background-color: rgba(208, 205, 208, 1);">VS Code 설치 방법 (Windows 기준)</mark></div>

1. **공식 사이트 접속**
    - https://code.visualstudio.com/ 에 접속


2. **설치 파일 다운로드**
    - **사용자의 운영체제를 자동으로 감지해서 가장 적절한 다운로드 버튼 보임** <br>
    - 메인 페이지에서 “Download for Windows” 버튼 클릭

        <img src="https://haesunbyun.github.io/common/images/comps/1pic.png" width=300>

3. **설치 프로그램 실행**
    - 다운로드한 설치 파일(VSCodeUserSetup-x64-x.x.x.exe 같은 이름)을 더블클릭해 실행

4. **설치 옵션 선택**
    - 환영 화면에서 "Next" 클릭
    
    - 라이선스 동의 체크 후 "Next"

        <img src="https://haesunbyun.github.io/common/images/comps/install1.png" width=300>    
    
    - 설치 경로 선택 (기본 경로 그대로 둬도 무방)

        <img src="https://haesunbyun.github.io/common/images/comps/install2.png" width=300>
        <img src="https://haesunbyun.github.io/common/images/comps/install3.png" width=300>   

    - <font color='red'>중요! 아래 옵션 체크하기</font>
        - **Add "Open with Code" action to Windows Explorer file context menu**
            - 윈도우 탐색기(파일 탐색기)에서 우클릭 메뉴에 "Code로 열기" 항목을 추가하는 옵션
        - **Add "Open with Code" action to Windows Explorer directory context menu**
            - 윈도우 탐색기에서 폴더(디렉터리)를 우클릭 메뉴에 "Open with Code" 항목을 추가하는 옵션
        - **Add to PATH**
            - 터미널에서 code 명령어를 쓸 수 있게 하는 옵션

            <img src="https://haesunbyun.github.io/common/images/comps/install4.png" width=300>

    - 옵션 체크 후 다음 → 설치

        <img src="https://haesunbyun.github.io/common/images/comps/install5.png" width=300>
        <img src="https://haesunbyun.github.io/common/images/comps/install6.png" width=300>

5. **설치 완료**
    - 설치가 완료되면 **"Launch Visual Studio Code"** 체크한 채로 종료(Finish) 클릭하면 바로 실행됨

        <img src="https://haesunbyun.github.io/common/images/comps/install7.png" width=300>    


## <div style="background-color:rgba(208, 205, 208, 1); padding: 10px; border-radius: 5px;"><mark style="background-color: rgba(208, 205, 208, 1);">VS Code 실행</mark></div>

- VS code를 처음 실행하면 아래와 같은 시작 화면이 보인다.

    <img src="https://haesunbyun.github.io/common/images/comps/install8.png" width=300>

- VS Code의 시작 화면인 "Get Started with vs code 페이지에 나오는 항목들은 VS Code를 사용자 맞춤형으로 설정하고, 유용한 기능을 빠르게 체험할 수 있도록 돕기 위한 것이다.
- 사용자 맞춤형 설정은 필요할 때 점진적으로 찾아 설정하면 된다.
    - **Use AI features with Copilot for free**: AI 코딩 도우미로, 코드 자동완성, 함수 작성, 설명 등을 도와준다.
    - **Choose your theme:** VS Code의 색상 테마(배경색, 글자색 등)를 설정한다.
    - **Rich support for all your languages**: 자주 사용하는 언어(예: Python, C++, JavaScript 등)에 맞는 확장 프로그램(Extension) 설치를 도와준다.
    - **Tune your settings**: VS Code의 기본 설정을 사용자 스타일에 맞게 조정할 수 있게 해주는 항목이다.
    - **Unlock productivity with the Command Palette**: 명령어, 설정 변경, 확장 설치, 깃 명령 등 거의 모든 작업을 이곳에서 수행 가능
    - **Watch video tutorials**: 공식 Microsoft 또는 커뮤니티에서 제공하는 기초 영상 가이드 링크로 연결(VS Code 사용법, 디버깅, Git 연동, 확장 기능 설치 등 주제로 짧은 영상들을 제공)

## <div style="background-color:rgba(208, 205, 208, 1); padding: 10px; border-radius: 5px;"><mark style="background-color: rgba(208, 205, 208, 1);">VS Code의 확장팩 추가 설치</mark></div>

VS Code는 가볍고 범용적인 에디터이기 때문에 Python 개발 환경을 갖추기 위해서는 몇 가지 **확장팩**(Extensions)을 추가로 설치해야 한다.

확장팩은 VS Code의 왼쪽 사이드에 네모 아이콘을 클릭하여 설치할 수 있다.

| 확장팩 이름                    | 설명                                                       |
| ------------------------- | -------------------------------------------------------- |
| **Python** (by Microsoft) | 파이썬 기본 확장. 문법 강조, 실행, 디버깅, 자동완성, 가상환경 등 핵심 기능 포함         |
| **Pylance**               | Python 코드를 더 잘 이해하고, 자동 완성이나 오류 체크를 빠르고 정확하게 해주는 언어 서버. 빠른 자동완성, 타입 힌트 등 고급 편집 기능.|
| **Jupyter**               | `.ipynb` 파일 실행을 원할 경우 필요 (예: streamlit과 함께 데이터 시각화 실험 등) |

이 세 가지는 Python 개발의 기본 세트이다.
