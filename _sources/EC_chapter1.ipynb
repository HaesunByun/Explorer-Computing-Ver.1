## <div style="background-color:rgba(208, 205, 208, 1); padding: 10px; border-radius: 5px;"><mark style="background-color: rgba(208, 205, 208, 1);">Anaconda 설치(Windows 기준)</mark></div>

- Anaconda는 파이썬(Python) 프로그래밍을 쉽게 사용할 수 있도록 여러 도구를 한 번에 묶어 놓은 통합 프로그램이다.
- 파이썬으로 데이터 분석, 인공지능, 시각화 등을 하려면 다양한 라이브러리가 필요한데, 이를 하나씩 설치하면 오류가 자주 발생한다. Anaconda는 이런 문제를 줄이기 위해 다음을 함께 제공한다.
    - Python (프로그래밍 언어)
    - 주요 라이브러리 (numpy, pandas, matplotlib 등)
    - Jupyter Notebook (코드 실습 도구)
    - conda (프로그램 환경 관리 도구)
- Anaconda는 <font color='red'>파이썬 실습에 필요한 기본 환경을 한 번에 설치해주는 패키지 세트</font>라고 보면 된다.

1. **공식 사이트 접속**
    - https://www.anaconda.com/download에 접속


2. **자신의 컴퓨터에 맞는 버전 선택**

3. **설치 프로그램 실행**
    - 다운로드한 설치 파일을 더블클릭해 실행

4. **설치 옵션 선택**
    1. 화면 ① Welcome ➡ Next 클릭
    2. 화면 ② License Agreement ➡ I Agree 클릭
    3. 화면 ③ Installation Type ➡ Just Me (Recommended) 선택✅ 
    4. 화면 ④ Installation Location
        - ⚠ Windows 사용자 계정 이름이 한글인 경우 오류가 발생할 수 있습니다.
        - 설치 경로를 기본값 대신 다음처럼 변경하세요: C:\anaconda3
        - 사용자 계정 이름이 영문이면 그대로 하셔도 됩니다.(예: C:\Users\Username\anaconda3)
    5. 화면 ⑤ Advanced Options (가장 중요)
        - <font color='red'>Add Anaconda to my PATH environment variable</font>
            - Windows에 이미 다른 Python이 설치되어 있는 경우 PATH가 꼬일 수 있으므로, Python이 설치되어 있는 경우에는 체크하지 마세요.
        - Register Anaconda as my default Python	✅ 체크 유지
    6. 설치 진행
    7. 설치 확인 (중요)
        - Windows 시작 메뉴에서 Anaconda Prompt 검색, 
        - 명령 입력 conda --version

## <div style="background-color:rgba(208, 205, 208, 1); padding: 10px; border-radius: 5px;"><mark style="background-color: rgba(208, 205, 208, 1);">Visual Studio Code 설치(Windows 기준)</mark></div>

- VS code는 마이크로소프트에서 개발한 **무료 오픈소스 코드 편집기**이다. 가볍지만 강력한 기능을 갖추고 있어 많은 개발자들이 사용하고 있다.

장점은 다음과 같다.<br>
- **가볍고 빠름**: 무거운 IDE(IntelliJ, Eclipse 등)보다 훨씬 가볍고 빠르게 실행된다.
- **강력한 확장성**: 다양한 확장(Extensions)을 통해 기능을 쉽게 추가할 수 있다. (예: Python Jupyter, Git 등)
- **다양한 언어 지원**: 기본적으로 많은 언어를 지원하고, 플러그인을 통해 거의 모든 프로그래밍 언어를 사용할 수 있다.
- **Git 통합**: Git이 내장되어 있어 별도 도구 없이도 버전 관리가 가능하다.
- **디버깅 도구 내장**: Node.js, Python 등 다양한 언어에 대한 디버깅을 VS Code 내에서 실행할 수 있어 개발 생산성이 높다.
- **멀티 플랫폼 지원**: Windows, macOS, Linux 모두 지원한다.

단점은 아래와 같다.<br>
- **초기 설정 필요**: 언어별 환경 세팅을 직접 설치하고 설정해야 해서 초보자에게는 약간 번거로울 수 있다.
- **리소스 사용 증가**: 여러 확장 기능을 많이 설치하면 속도가 느려지고 메모리 사용량이 증가할 수 있다.
- **GUI 기반 작업엔 불편**: GUI 앱을 설계할 때 드래그앤드롭 방식의 UI 빌더가 지원되지 않아 효율적이지 않다.

1. **공식 사이트 접속**
    - https://code.visualstudio.com/ 에 접속


2. **설치 파일 다운로드**
    - **사용자의 운영체제를 자동으로 감지해서 가장 적절한 다운로드 버튼 보임** <br>

## <div style="background-color:rgba(208, 205, 208, 1); padding: 10px; border-radius: 5px;"><mark style="background-color: rgba(208, 205, 208, 1);">VS Code의 확장팩 추가 설치</mark></div>

VS Code는 가볍고 범용적인 에디터이기 때문에 Python 개발 환경을 갖추기 위해서는 몇 가지 **확장팩**(Extensions)을 추가로 설치해야 한다.

확장팩은 VS Code의 왼쪽 사이드에 네모 아이콘(Extensions)을 클릭하여 설치할 수 있다.

| 확장팩 이름                    | 설명                                                       |
| ------------------------- | -------------------------------------------------------- |
| **Python** (by Microsoft) | 파이썬 기본 확장. 문법 강조, 실행, 디버깅, 자동완성, 가상환경 등 핵심 기능 포함         |
| **Pylance**               | Python 코드를 더 잘 이해하고, 자동 완성이나 오류 체크를 빠르고 정확하게 해주는 언어 서버. 빠른 자동완성, 타입 힌트 등 고급 편집 기능.|
| **Jupyter**               | `.ipynb` 파일 실행을 원할 경우 필요 (예: streamlit과 함께 데이터 시각화 실험 등) |

이 세 가지는 Python 개발의 기본 세트이다.

## <div style="background-color:rgba(208, 205, 208, 1); padding: 10px; border-radius: 5px;"><mark style="background-color: rgba(208, 205, 208, 1);">가상 환경 구축 </mark></div>

- 가상환경이란 하나의 컴퓨터 안에서 독립적인 Python 실행 공간을 만드는 것을 의미한다.
- 가상환경을 사용하면 하나의 컴퓨터 안에서 서로 다른 Python 버전을 각각 독립적으로 사용할 수 있다.
- 또한 각 환경마다 필요한 라이브러리들을 별도로 설치하여 관리할 수 있다. 
- 한 환경에서 설치하거나 변경한 내용이 다른 환경에 영향을 주지 않는다.
- 터미널을 열어 아래 명령어를 실행해보자.
- 아래 코드는 computing이라는 이름의 독립적인 Python 개발 환경을 생성한다. 이때 특정 Python 버전을 지정하지 않으면 Conda가 현재 설치 가능한 적절한 Python 버전을 선택한다.

```python
conda create -n computing python
```

- 가상환경이 만들어지면 아래 명령어로 활성화 시킨다.
- 정상 활성화가 되면 프롬프트가 (computing) C:\Users\... 와 같이 바뀔 것이다.

```python
conda activate computing
```

- 정상 활성화가 되면 프롬프트가 (computing) C:\Users\... 와 같이 바뀔 것이다.
- 이제 모든 환경 설정이 완료되었다.
- 우리 수업에서는 VS Code에서 인터프리터를 선택할 때 Anaconda3\envs\computing\python.exe와 같이 (computing) 환경이 표시된 항목을 선택하면 된다.
